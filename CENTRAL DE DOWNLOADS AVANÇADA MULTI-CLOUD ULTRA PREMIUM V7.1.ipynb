```python
#@title <b><img src="https://cdn-icons-png.flaticon.com/512/11271/11271573.png" width="22" height="22" style="vertical-align: middle; margin-right: 8px;"> CENTRAL MULTI-CLOUD EXPLORER</b> { display-mode: "form" }
import os, ipywidgets as widgets, sys, subprocess, shutil, time
from IPython.display import display, clear_output

estilo_matrix = widgets.HTML("""
<style>
    .output_subarea, .output_wrapper, .jupyter-widgets { background-color: #000000 !important; font-family: 'Courier New', monospace !important; }
    .widget-label { color: #00FF00 !important; font-family: 'Courier New', monospace; font-weight: bold; min-width: 90px !important; display: inline-block; }
    .matrix-title { color: #00FF00; font-family: 'Courier New', monospace; font-weight: bold; font-size: 14px; text-shadow: 0 0 5px #00FF00; }
    .widget-dropdown select { background-color: #000000 !important; color: #00FF00 !important; border: 2px solid #00FF00 !important; font-family: 'Courier New', monospace; font-weight: bold; }
    .widget-text input { background-color: #000000 !important; color: #00FF00 !important; border: 2px solid #00FF00 !important; font-family: 'Courier New', monospace; font-weight: bold; }

    /* CORES DE ALTO CONTRASTE PARA OS BOTÕES SOLICITADAS PELO MATTOS */
    .btn-inj-matrix { background-color: #008f39 !important; color: #FFFFFF !important; font-weight: bold !important; border: 2px solid #00ff00 !important; }
    .btn-des-matrix { background-color: #990000 !important; color: #FFFFFF !important; font-weight: bold !important; border: 2px solid #ff0000 !important; }
    .btn-disp-matrix { background-color: #00ff00 !important; color: #000000 !important; font-family: 'Courier New', monospace !important; font-weight: bold !important; font-size: 13px !important; border: 3px solid #ffffff !important; box-shadow: 0 0 10px #00ff00 !important; }
</style>
""")
display(estilo_matrix)

def obter_remotes():
    try:
        resultado = os.popen('rclone listremotes').read().splitlines()
        lista = [r.rstrip(':') for r in resultado if r.strip()]
        if 'GoogleDrive (Nativo)' not in lista: lista.append('GoogleDrive (Nativo)')
        return lista
    except: return ['DROPBOX-BACKUPGERAL', 'pCloud-FMC2', 'onedrive', 'GoogleDrive (Nativo)']

lista_contas = obter_remotes()
lbl_titulo = widgets.HTML("<div class='matrix-title'>🟢 CENTRAL AVANÇADA MATRIX — CONTROLE SELETIVO DE INTERFACE</div><br>")
drp_nuvem_alvo = widgets.Dropdown(options=lista_contas, description='☁️ Injetar:')

btn_injetar = widgets.Button(description='🟢 INJETAR CONEXÃO NA BARRA LATERAL', layout=widgets.Layout(width='48%', height='40px'))
btn_injetar.add_class('btn-inj-matrix')

btn_desinjetar = widgets.Button(description='❌ DESINJETAR / LIMPAR DA LATERAL', layout=widgets.Layout(width='48%', height='40px'))
btn_desinjetar.add_class('btn-des-matrix')

txt_origem_cmd = widgets.Text(placeholder='Copie o caminho na barra EXPLORER à esquerda e cole aqui', description='🎯 Origem:', layout=widgets.Layout(width='95%'))
txt_destino_cmd = widgets.Text(placeholder='Copie o caminho na barra EXPLORER à esquerda e cole aqui', description='📍 Destino:', layout=widgets.Layout(width='95%'))

btn_executar_copia = widgets.Button(description='⚡ DISPARAR TRANSFERÊNCIA ULTRAVELOZ ENTRE AS CONTAS', layout=widgets.Layout(width='95%', height='45px'))
btn_executar_copia.add_class('btn-disp-matrix')

output_painel = widgets.Output(layout=widgets.Layout(width='95%', max_height='300px', border='1px solid #333', overflow_y='auto'))

def atualizar_campos_texto(change):
    nuvem_nome = drp_nuvem_alvo.value
    txt_origem_cmd.value = "GoogleDrive:" if 'Nativo' in nuvem_nome else f"{nuvem_nome}:"
    txt_destino_cmd.value = "GoogleDrive:" if 'Nativo' in nuvem_nome else f"{nuvem_nome}:"
drp_nuvem_alvo.observe(atualizar_campos_texto, names='value')

def processar_injecao(b):
    with output_painel:
        clear_output(); nuvem = drp_nuvem_alvo.value; pasta_alvo = f"/content/{nuvem}"
        if 'Nativo' in nuvem:
            print("🤖 Inicializando barramento nativo isolado para o Google Drive...")
            try:
                from google.colab import drive
                drive.mount('/content/GoogleDrive', force_remount=True)
                print("🟩 MATRIX INJECTED! Dê um clique no botão de atualizar (🔄) da barra lateral.")
            except Exception as e: print(f"❌ Erro ao vincular o Google Drive: {e}")
        else:
            if not os.path.exists(pasta_alvo): os.makedirs(pasta_alvo, exist_ok=True)
            print(f"🟩 Inicializando montagem estável para: {nuvem}...")
            os.system(f"fusermount -u {pasta_alvo} >/dev/null 2>&1; pkill -f 'mount {nuvem}:'")
            cmd = f'rclone mount "{nuvem}:" "{pasta_alvo}" --vfs-cache-mode writes --allow-other --daemon --ignore-errors &'
            subprocess.Popen(cmd, shell=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
            time.sleep(3)
            print("🟩 MATRIX INJECTED! Dê um clique no botão de atualizar (🔄) da barra lateral.")

def processar_desinjecao(b):
    with output_painel:
        clear_output(); nuvem = drp_nuvem_alvo.value
        if 'Nativo' in nuvem:
            os.system("fusermount -u /content/GoogleDrive >/dev/null 2>&1")
            print("🗑️ Concluído! O Google Drive foi removido da barra lateral.")
        else:
            pasta_alvo = f"/content/{nuvem}"
            print(f"🛑 Desmontando e limpando conexões de: {nuvem}...")
            os.system(f"fusermount -u {pasta_alvo} >/dev/null 2>&1; pkill -f 'mount {nuvem}:'")
            if os.path.exists(pasta_alvo):
                try: shutil.rmtree(pasta_alvo)
                except: pass
            print(f"🗑️ Concluído! A pasta '{nuvem}' foi totalmente removida.")

def executar_rclone_direto(b):
    with output_painel:
        clear_output(); origem, destino = txt_origem_cmd.value.strip(), txt_destino_cmd.value.strip()
        if not origem or not destino: print("❌ [ERRO] Preencha os campos de Origem e Destino!"); return
        print(f"\n=============================================\n⚡ EXECUTANDO CÓPIA DIRETA ENTRE AS CONTAS\n=============================================\n")
        cmd = f'rclone copy "{origem}" "{destino}" -P --stats 1s --stats-one-line --fast-list --ignore-errors' if (origem.endswith('/') or '.' not in os.path.basename(origem)) else f'rclone copyto "{origem}" "{destino}/{os.path.basename(origem)}" -P --stats 1s --stats-one-line --fast-list --ignore-errors'
        processo = subprocess.Popen(cmd, shell=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)
        for linha in iter(processo.stdout.readline, ''): sys.stdout.write(linha); sys.stdout.flush()
        processo.stdout.close(); processo.wait(); print("\n🟩 TRANSFERÊNCIA FINALIZADA COM SUCESSO!")

btn_injetar.on_click(processar_injecao); btn_desinjetar.on_click(processar_desinjecao); btn_executar_copia.on_click(executar_rclone_direto)

display(widgets.VBox([
    lbl_titulo, widgets.HTML("<hr style='border: 1px solid #00FF00;'>"),
    drp_nuvem_alvo, widgets.HTML("<br>"),
    widgets.HBox([btn_injetar, btn_desinjetar], layout=widgets.Layout(justify_content='space-between', width='95%')),
    widgets.HTML("<br><hr style='border: 1px dashed #00FF00;'><b style='color: #00FF00; font-family: monospace;'>⚡ TERMINAL DE TRANSFERÊNCIA DIRETA:</b><br>"),
    txt_origem_cmd, txt_destino_cmd, btn_executar_copia, output_painel
]))
```
