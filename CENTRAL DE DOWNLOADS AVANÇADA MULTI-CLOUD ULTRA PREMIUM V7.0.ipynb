```python
#@title <b><img src="https://avatars.githubusercontent.com/u/24937341?s=200&v=4" width="22" height="22" style="vertical-align: middle; margin-right: 8px;"> RCLONE WEB GUI</b> { display-mode: "form" }

import os
import sys
import time
import subprocess
import ipywidgets as widgets

from IPython.display import display, clear_output


# Injeção estática Matrix Premium calibrada no Verde Padrão
estilo_matrix_gui = widgets.HTML("""
<style>
.output_subarea, .output_wrapper, .jupyter-widgets {
    background-color: #000000 !important;
    font-family: 'Courier New', monospace !important;
}

.matrix-title-gui {
    color: #FFFFFF !important;
    font-family: 'Courier New', monospace;
    font-weight: bold;
    font-size: 14px;
    text-shadow: 0 0 5px #FFFFFF;
}

.widget-label {
    color: #00FF00 !important;
    font-family: 'Courier New', monospace;
    font-weight: bold;
    min-width: 140px !important;
    display: inline-block;
}

.widget-dropdown select, .widget-text input {
    background-color: #000000 !important;
    color: #00FF00 !important;
    border: 2px solid #00FF00 !important;
    font-family: 'Courier New', monospace;
    font-weight: bold;
}

.matrix-btn-gui {
    background-color: #005f00 !important;
    color: #00FF00 !important;
    border: 2px solid #00FF00 !important;
    font-family: 'Courier New', monospace !important;
    font-weight: bold !important;
    text-shadow: 0 0 5px #00FF00;
}
</style>

<div class="matrix-title-gui">
🎛️ CENTRAL GRAPHICAL MATRIX — RELAY TÚNEL INTERATIVO
</div>

<hr style="border: 1px dashed #00FF00;">
""")

display(estilo_matrix_gui)


# Caixas gráficas de seleção e parametrização
drp_tipo_tunel = widgets.Dropdown(
    options=[
        ('1 - LocalTunnel (Padrão)', 'localtunnel'),
        ('2 - Cloudflare Tunnel (Estabilizado)', 'cloudflare'),
        ('3 - Ngrok Tunnel (Requer Token)', 'ngrok'),
        ('4 - Google Colab Secure Link', 'colab')
    ],
    value='localtunnel',
    description='📡 Tipo de Túnel:',
    layout=widgets.Layout(width='95%')
)


txt_ngrok_token = widgets.Text(
    placeholder='🔑 Opcional: Cole seu Authtoken do Ngrok aqui apenas se escolher a Opção 3...',
    description='🔑 Token Ngrok:',
    layout=widgets.Layout(width='95%')
)


btn_ativar_gui_cell = widgets.Button(
    description='🌐 LIGAR RELAY INTERFACE GRÁFICA (RCLONE WEB GUI)',
    layout=widgets.Layout(width='95%', height='45px')
)

btn_ativar_gui_cell.add_class('matrix-btn-gui')


output_console_gui = widgets.Output(
    layout=widgets.Layout(
        width='95%',
        max_height='250px',
        border='1px solid #333',
        overflow_y='auto'
    )
)


# ==============================================================
# FUNÇÃO DE LIMPEZA DOS TÚNEIS
# ==============================================================
def limpar_tuneis_anteriores():

    processos = [
        "lt --port 5572",
        "cloudflared tunnel",
        "ngrok http 5572"
    ]

    for processo in processos:
        try:
            subprocess.run(
                ["pkill", "-f", processo],
                stdout=subprocess.DEVNULL,
                stderr=subprocess.DEVNULL
            )
        except Exception:
            pass

    time.sleep(1)


# ==============================================================
# OPÇÃO 1 - LOCALTUNNEL
# ==============================================================
def iniciar_localtunnel():

    print("📡 Sincronizando canais de rede via LocalTunnel...")

    # Instala somente se ainda não existir
    if subprocess.run(
        "command -v lt",
        shell=True,
        stdout=subprocess.DEVNULL,
        stderr=subprocess.DEVNULL
    ).returncode != 0:

        print("📦 Instalando LocalTunnel...")
        resultado = subprocess.run(
            "npm install -g localtunnel",
            shell=True,
            stdout=subprocess.DEVNULL,
            stderr=subprocess.DEVNULL
        )

        if resultado.returncode != 0:
            print("❌ Falha ao instalar o LocalTunnel.")
            return

    print("⏳ Criando túnel LocalTunnel...")

    processo_tunnel = subprocess.Popen(
        "lt --port 5572",
        shell=True,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1
    )

    inicio = time.time()

    while time.time() - inicio < 40:

        linha_log = processo_tunnel.stdout.readline()

        if linha_log:

            linha_log = linha_log.strip()

            if "url is:" in linha_log:

                link_localtunnel = linha_log.split(
                    "url is:"
                )[-1].strip()

                print(
                    f"🔗 LINK GERADO COM SUCESSO -> {link_localtunnel}"
                )

                return link_localtunnel

            if "https://" in linha_log:

                partes = linha_log.split("https://")

                for parte in partes[1:]:

                    if "loca.lt" in parte:

                        link_localtunnel = (
                            "https://" +
                            parte.split()[0].strip()
                        )

                        print(
                            f"🔗 LINK GERADO COM SUCESSO -> {link_localtunnel}"
                        )

                        return link_localtunnel

        time.sleep(0.2)

    print(
        "⚠️ O LocalTunnel demorou para fornecer o endereço."
    )

    return None


# ==============================================================
# OPÇÃO 2 - CLOUDFLARE TUNNEL
# ==============================================================
def iniciar_cloudflare():

    print(
        "☁️ Vinculando barramento oficial e estável da Cloudflare..."
    )

    caminho_cloudflared = "/usr/local/bin/cloudflared"

    # ----------------------------------------------------------
    # Verifica se o executável já existe
    # ----------------------------------------------------------
    if not os.path.exists(caminho_cloudflared):

        print("📦 Instalando Cloudflared...")

        arquitetura = subprocess.check_output(
            "dpkg --print-architecture",
            shell=True,
            text=True
        ).strip()

        if arquitetura == "amd64":
            arquivo_cf = "cloudflared-linux-amd64"
        elif arquitetura == "arm64":
            arquivo_cf = "cloudflared-linux-arm64"
        elif arquitetura == "armhf":
            arquivo_cf = "cloudflared-linux-arm"
        else:
            arquivo_cf = "cloudflared-linux-amd64"

        url_download = (
            "https://github.com/cloudflare/cloudflared/"
            f"releases/latest/download/{arquivo_cf}"
        )

        resultado = subprocess.run(
            [
                "curl",
                "-L",
                "--fail",
                "--retry",
                "3",
                "--connect-timeout",
                "15",
                "-o",
                caminho_cloudflared,
                url_download
            ],
            stdout=subprocess.DEVNULL,
            stderr=subprocess.DEVNULL
        )

        if resultado.returncode != 0:

            print(
                "⚠️ Download direto falhou. Tentando instalação pelo repositório Cloudflare..."
            )

            subprocess.run(
                "curl -fsSL https://pkg.cloudflare.com/cloudflare-main.gpg "
                "| tee /usr/share/keyrings/cloudflare-main.gpg >/dev/null",
                shell=True,
                stdout=subprocess.DEVNULL,
                stderr=subprocess.DEVNULL
            )

            subprocess.run(
                "echo 'deb [signed-by=/usr/share/keyrings/cloudflare-main.gpg] "
                "https://pkg.cloudflare.com/cloudflared any main' "
                "| tee /etc/apt/sources.list.d/cloudflared.list >/dev/null",
                shell=True,
                stdout=subprocess.DEVNULL,
                stderr=subprocess.DEVNULL
            )

            subprocess.run(
                "apt-get update -q",
                shell=True,
                stdout=subprocess.DEVNULL,
                stderr=subprocess.DEVNULL
            )

            resultado_apt = subprocess.run(
                "apt-get install -y cloudflared -q",
                shell=True,
                stdout=subprocess.DEVNULL,
                stderr=subprocess.DEVNULL
            )

            if resultado_apt.returncode != 0:

                print(
                    "❌ Não foi possível instalar o Cloudflared."
                )

                return None

            if os.path.exists("/usr/bin/cloudflared"):

                caminho_cloudflared = "/usr/bin/cloudflared"

        else:

            subprocess.run(
                f"chmod +x {caminho_cloudflared}",
                shell=True,
                stdout=subprocess.DEVNULL,
                stderr=subprocess.DEVNULL
            )

    # ----------------------------------------------------------
    # Verificação final
    # ----------------------------------------------------------
    if not os.path.exists(caminho_cloudflared):

        print(
            "❌ Executável cloudflared não encontrado."
        )

        return None

    print(
        "⏳ Criando Quick Tunnel Cloudflare..."
    )

    log_cf = "/content/cf_gui.log"

    try:
        os.remove(log_cf)
    except Exception:
        pass

    processo_cf = subprocess.Popen(
        [
            caminho_cloudflared,
            "tunnel",
            "--url",
            "http://127.0.0.1:5572",
            "--no-autoupdate"
        ],
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1
    )

    inicio = time.time()

    link_cloudflare = None

    while time.time() - inicio < 45:

        linha = processo_cf.stdout.readline()

        if linha:

            linha = linha.strip()

            if "trycloudflare.com" in linha:

                encontrados = []

                for parte in linha.split():

                    if "https://" in parte and "trycloudflare.com" in parte:

                        encontrados.append(
                            parte.strip()
                            .replace('"', '')
                            .replace("'", "")
                            .replace(",", "")
                        )

                if encontrados:

                    link_cloudflare = encontrados[0]

                    print(
                        f"🔗 LINK GERADO COM SUCESSO -> {link_cloudflare}"
                    )

                    return link_cloudflare

        time.sleep(0.2)

    # ----------------------------------------------------------
    # Segunda tentativa através do log
    # ----------------------------------------------------------
    try:

        linhas_restantes = processo_cf.stdout.read()

        if linhas_restantes:

            for parte in linhas_restantes.split():

                if (
                    "https://" in parte
                    and "trycloudflare.com" in parte
                ):

                    link_cloudflare = (
                        parte.strip()
                        .replace('"', '')
                        .replace("'", "")
                        .replace(",", "")
                    )

                    print(
                        f"🔗 LINK GERADO COM SUCESSO -> {link_cloudflare}"
                    )

                    return link_cloudflare

    except Exception:
        pass

    print(
        "⚠️ O Cloudflare não forneceu o endereço público dentro do tempo esperado."
    )

    print(
        "👉 Tente clicar novamente no botão verde."
    )

    return None


# ==============================================================
# OPÇÃO 3 - NGROK
# ==============================================================
def iniciar_ngrok(token_inserido):

    if not token_inserido:

        print(
            "❌ [ERRO] Para rodar a rota do Ngrok, "
            "você precisa colar o seu Authtoken no campo "
            "'Token Ngrok' acima!"
        )

        print(
            "👉 Pegue seu token grátis no site oficial do Ngrok."
        )

        return None

    print(
        "🚀 Autenticando credenciais no gateway Ngrok..."
    )

    # ----------------------------------------------------------
    # Verifica instalação
    # ----------------------------------------------------------
    if subprocess.run(
        "command -v ngrok",
        shell=True,
        stdout=subprocess.DEVNULL,
        stderr=subprocess.DEVNULL
    ).returncode != 0:

        print(
            "📦 Instalando Ngrok..."
        )

        caminho_ngrok = "/usr/local/bin/ngrok"

        download_ngrok = (
            "curl -fsSL "
            "https://bin.ngrok.com/c/bNyj1mQV4c/"
            "ngrok-v3-stable-linux-amd64.tgz "
            "| tar -xz -C /tmp"
        )

        resultado = subprocess.run(
            download_ngrok,
            shell=True,
            stdout=subprocess.DEVNULL,
            stderr=subprocess.DEVNULL
        )

        if resultado.returncode != 0:

            # Método alternativo oficial via APT
            print(
                "⚠️ Download direto falhou. Tentando instalação via APT..."
            )

            subprocess.run(
                "curl -sSL https://ngrok-agent.s3.amazonaws.com/ngrok.asc "
                "| tee /etc/apt/trusted.gpg.d/ngrok.asc >/dev/null",
                shell=True,
                stdout=subprocess.DEVNULL,
                stderr=subprocess.DEVNULL
            )

            subprocess.run(
                "echo 'deb https://ngrok-agent.s3.amazonaws.com bookworm main' "
                "| tee /etc/apt/sources.list.d/ngrok.list >/dev/null",
                shell=True,
                stdout=subprocess.DEVNULL,
                stderr=subprocess.DEVNULL
            )

            subprocess.run(
                "apt-get update -q",
                shell=True,
                stdout=subprocess.DEVNULL,
                stderr=subprocess.DEVNULL
            )

            resultado_apt = subprocess.run(
                "apt-get install -y ngrok -q",
                shell=True,
                stdout=subprocess.DEVNULL,
                stderr=subprocess.DEVNULL
            )

            if resultado_apt.returncode != 0:

                print(
                    "❌ Não foi possível instalar o Ngrok."
                )

                return None

        else:

            if os.path.exists("/tmp/ngrok"):

                subprocess.run(
                    f"mv /tmp/ngrok {caminho_ngrok}",
                    shell=True,
                    stdout=subprocess.DEVNULL,
                    stderr=subprocess.DEVNULL
                )

                subprocess.run(
                    f"chmod +x {caminho_ngrok}",
                    shell=True,
                    stdout=subprocess.DEVNULL,
                    stderr=subprocess.DEVNULL
                )

    # ----------------------------------------------------------
    # Procura o executável
    # ----------------------------------------------------------
    caminho_ngrok = "ngrok"

    if os.path.exists("/usr/local/bin/ngrok"):
        caminho_ngrok = "/usr/local/bin/ngrok"

    # ----------------------------------------------------------
    # Configura Authtoken
    # ----------------------------------------------------------
    resultado_token = subprocess.run(
        [
            caminho_ngrok,
            "config",
            "add-authtoken",
            token_inserido
        ],
        stdout=subprocess.DEVNULL,
        stderr=subprocess.PIPE,
        text=True
    )

    if resultado_token.returncode != 0:

        print(
            "❌ Falha ao configurar o Authtoken do Ngrok."
        )

        print(
            "⚠️ Verifique se o token foi copiado corretamente."
        )

        return None

    # ----------------------------------------------------------
    # Inicia Ngrok
    # ----------------------------------------------------------
    print(
        "⏳ Abrindo túnel Ngrok..."
    )

    processo_ngrok = subprocess.Popen(
        [
            caminho_ngrok,
            "http",
            "5572"
        ],
        stdout=subprocess.DEVNULL,
        stderr=subprocess.DEVNULL
    )

    # ----------------------------------------------------------
    # Aguarda API local do Ngrok
    # API correta: 127.0.0.1:4040
    # ----------------------------------------------------------
    import urllib.request
    import json

    inicio = time.time()

    while time.time() - inicio < 30:

        try:

            res = urllib.request.urlopen(
                "http://127.0.0.1:4040/api/tunnels",
                timeout=3
            )

            dados = json.loads(
                res.read().decode()
            )

            for tunel in dados.get("tunnels", []):

                url_ngrok = tunel.get("public_url", "")

                if url_ngrok.startswith("https://"):

                    print(
                        f"🔗 LINK GERADO COM SUCESSO -> {url_ngrok}"
                    )

                    return url_ngrok

        except Exception:
            pass

        time.sleep(1)

    print(
        "⚠️ Falha ao obter o link público do Ngrok."
    )

    print(
        "👉 Verifique se o Authtoken é válido."
    )

    return None


# ==============================================================
# OPÇÃO 4 - GOOGLE COLAB SECURE LINK
# ==============================================================
def iniciar_colab_secure_link():

    print(
        "🟦 Criando Google Colab Secure Link..."
    )

    try:

        from google.colab import output

        url_colab = output.eval_js(
            "google.colab.kernel.proxyPort(5572)"
        )

        if url_colab:

            print(
                f"🔗 LINK GERADO COM SUCESSO -> {url_colab}"
            )

            return url_colab

    except Exception as e:

        print(
            "❌ Não foi possível criar o Secure Link do Google Colab."
        )

        print(
            f"Detalhes: {e}"
        )

    return None


# ==============================================================
# MOTOR PRINCIPAL
# ==============================================================
def disparar_rclone_web_gui_isolado(b):

    with output_console_gui:

        clear_output()

        tipo_conexao = drp_tipo_tunel.value
        token_inserido = txt_ngrok_token.value.strip()

        print(
            "💻 Inicializando Servidor Interno Rclone Web GUI..."
        )

        # ------------------------------------------------------
        # Limpeza dos túneis anteriores
        # ------------------------------------------------------
        limpar_tuneis_anteriores()

        # ------------------------------------------------------
        # Garante que o Rclone anterior seja encerrado
        # ------------------------------------------------------
        try:

            subprocess.run(
                ["pkill", "-f", "rclone rcd"],
                stdout=subprocess.DEVNULL,
                stderr=subprocess.DEVNULL
            )

        except Exception:
            pass

        time.sleep(1)

        # ------------------------------------------------------
        # Ativa o motor da GUI com autenticação firme original
        # ------------------------------------------------------
        cmd_gui = (
            "rclone rcd "
            "--rc-web-gui "
            "--rc-user=matrix "
            "--rc-pass=matrix "
            "--rc-addr=127.0.0.1:5572 "
            "--rc-no-auth=false "
            "--rc-web-gui-no-open-browser &"
        )

        processo_gui = subprocess.Popen(
            cmd_gui,
            shell=True,
            stdout=subprocess.DEVNULL,
            stderr=subprocess.DEVNULL
        )

        time.sleep(2)

        # ------------------------------------------------------
        # Verifica se o Rclone realmente iniciou
        # ------------------------------------------------------
        porta_ok = subprocess.run(
            "curl -s --max-time 3 http://127.0.0.1:5572 >/dev/null",
            shell=True
        ).returncode == 0

        if not porta_ok:

            print(
                "❌ O Rclone Web GUI não respondeu na porta 5572."
            )

            print(
                "⚠️ Aguarde alguns segundos e tente novamente."
            )

            return

        print(
            "\n=========================================================="
        )

        print(
            "🟩 MATRIX GUI ACTIVATED! INICIALIZANDO ROTEAMENTO:"
        )

        print(
            "==========================================================\n"
        )

        print(
            "👉 Usuário: matrix"
        )

        print(
            "👉 Senha: matrix\n"
        )

        # ======================================================
        # OPÇÃO 1 - LOCALTUNNEL
        # ======================================================
        if tipo_conexao == 'localtunnel':

            iniciar_localtunnel()

        # ======================================================
        # OPÇÃO 2 - CLOUDFLARE
        # ======================================================
        elif tipo_conexao == 'cloudflare':

            iniciar_cloudflare()

        # ======================================================
        # OPÇÃO 3 - NGROK
        # ======================================================
        elif tipo_conexao == 'ngrok':

            iniciar_ngrok(token_inserido)

        # ======================================================
        # OPÇÃO 4 - GOOGLE COLAB SECURE LINK
        # ======================================================
        elif tipo_conexao == 'colab':

            iniciar_colab_secure_link()

        print(
            "\n=========================================================="
        )


# ==============================================================
# ATIVAÇÃO DO BOTÃO
# ==============================================================
btn_ativar_gui_cell.on_click(
    disparar_rclone_web_gui_isolado
)


display(
    widgets.VBox([
        drp_tipo_tunel,
        txt_ngrok_token,
        widgets.HTML("<br>"),
        btn_ativar_gui_cell,
        output_console_gui
    ])
)
```

```python
#@title <b><img src="https://cdn.jsdelivr.net/gh/homarr-labs/dashboard-icons/png/rclone.png" width="22" height="22" style="vertical-align: middle; margin-right: 8px;"> RCLONE LAB</b> { display-mode: "form" }

import os, psutil, time, subprocess, shlex
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML


# ================================================================
# INJEÇÃO ESTÁTICA MATRIX PREMIUM
# ================================================================

estilo_matrix_gui = widgets.HTML("""
<style>
    .output_subarea, .output_wrapper, .jupyter-widgets { background-color: #000000 !important; font-family: 'Courier New', monospace !important; }
    .matrix-title-gui { color: #FFFFFF !important; font-family: 'Courier New', monospace; font-weight: bold; font-size: 14px; text-shadow: 0 0 5px #FFFFFF; }
    .matrix-subtitle-gui { color: #00FF00 !important; font-family: 'Courier New', monospace; font-weight: bold; font-size: 12px; margin-top: 15px; margin-bottom: 5px; text-shadow: 0 0 3px #00FF00; text-transform: uppercase; }
    .widget-label { color: #00FF00 !important; font-family: 'Courier New', monospace; font-weight: bold; min-width: 180px !important; display: inline-block; }
    .widget-dropdown select, .widget-text input, .widget-checkbox input { background-color: #000000 !important; color: #00FF00 !important; border: 2px solid #00FF00 !important; font-family: 'Courier New', monospace; font-weight: bold; }
    .widget-checkbox { color: #00FF00 !important; font-family: 'Courier New', monospace; font-weight: bold; }
    .widget-readout { color: #00FF00 !important; font-family: 'Courier New', monospace !important; font-weight: bold; }
    .matrix-btn-gui { background-color: #005f00 !important; color: #00FF00 !important; border: 2px solid #00FF00 !important; font-family: 'Courier New', monospace !important; font-weight: bold !important; text-shadow: 0 0 5px #00FF00; }
</style>
<div class="matrix-title-gui">🎛️ CENTRAL GRAPHICAL MATRIX — CONTROLADOR OPERACIONAL RCLONELAB</div>
<hr style="border: 1px dashed #00FF00;">
""")

display(estilo_matrix_gui)


# ================================================================
# 1. COMPONENTES INTERATIVOS DE CONTROLE (UI)
# ================================================================

lbl_diretrizes = widgets.HTML(
    '<div class="matrix-subtitle-gui">📍 DIRETRIZES DA MISSÃO ATUAL</div>'
)

drp_mode = widgets.Dropdown(
    options=[
        "Config",
        "Copy",
        "Move",
        "Link",
        "Sync",
        "Checker",
        "Deduplicate",
        "Remove Empty Directories",
        "Empty Trash",
        "qBittorrent"
    ],
    value="Copy",
    description="🛠️ Modo Operacional:",
    layout=widgets.Layout(width='95%')
)

drp_compare = widgets.Dropdown(
    options=[
        "Size & Mod-Time",
        "Size & Checksum",
        "Only Mod-Time",
        "Only Size",
        "Only Checksum"
    ],
    value="Size & Mod-Time",
    description="🔍 Comparação:",
    layout=widgets.Layout(width='95%')
)

txt_source = widgets.Text(
    placeholder="Ex: drive:Origem ou /content/pasta",
    description="📂 Origem (Source):",
    layout=widgets.Layout(width='95%')
)

txt_destination = widgets.Text(
    placeholder="Ex: drive:Destino ou /content/destino",
    description="📁 Destino (Destination):",
    layout=widgets.Layout(width='95%')
)

sld_transfers = widgets.IntSlider(
    value=4,
    min=1,
    max=20,
    step=1,
    description="⚡ Transferências:",
    layout=widgets.Layout(width='95%')
)

sld_checkers = widgets.IntSlider(
    value=4,
    min=1,
    max=20,
    step=1,
    description="🛡️ Verificadores:",
    layout=widgets.Layout(width='95%')
)


lbl_globais = widgets.HTML(
    '<div class="matrix-subtitle-gui">⚙️ CONFIGURAÇÃO GLOBAL DA MALHA</div>'
)

chk_simple_output = widgets.Checkbox(
    value=True,
    description="📟 Saída de Log Simplificada no Terminal",
    layout=widgets.Layout(width='95%')
)

chk_skip_newer = widgets.Checkbox(
    value=False,
    description="⏭️ Pular arquivos mais recentes no destino",
    layout=widgets.Layout(width='95%')
)

chk_skip_exist = widgets.Checkbox(
    value=False,
    description="🚫 Pular todos os arquivos já existentes",
    layout=widgets.Layout(width='95%')
)

chk_no_update_mod = widgets.Checkbox(
    value=False,
    description="⏰ Não atualizar Mod-Time se idênticos",
    layout=widgets.Layout(width='95%')
)

chk_gdrive_opt = widgets.Checkbox(
    value=False,
    description="🤖 Otimização específica para Google Drive",
    layout=widgets.Layout(width='95%')
)

chk_gdrive_server_side = widgets.Checkbox(
    value=False,
    description="🧬 Cópia Server-Side (Entre Contas GDrive)",
    layout=widgets.Layout(width='95%')
)

chk_dry_run = widgets.Checkbox(
    value=False,
    description="🧪 Dry Run (Simulação sem alterações reais)",
    layout=widgets.Layout(width='95%')
)

drp_log_file = widgets.Dropdown(
    options=["OFF", "NOTICE", "INFO", "ERROR", "DEBUG"],
    value="OFF",
    description="📝 Gravação de Log:",
    layout=widgets.Layout(width='95%')
)

txt_extra_args = widgets.Text(
    placeholder="--exclude *.tmp --retries 3",
    description="➕ Argumentos Extra:",
    layout=widgets.Layout(width='95%')
)


lbl_sync = widgets.HTML(
    '<div class="matrix-subtitle-gui">↪️ CONFIGURAÇÕES DE SINCRONIZAÇÃO (SYNC)</div>'
)

drp_sync_mode = widgets.Dropdown(
    options=[
        "Delete during transfer",
        "Delete before transfering",
        "Delete after transfering"
    ],
    value="Delete during transfer",
    description="🔄 Modo de Exclusão:",
    layout=widgets.Layout(width='95%')
)

chk_track_renames = widgets.Checkbox(
    value=False,
    description="🏷️ Rastrear arquivos renomeados",
    layout=widgets.Layout(width='95%')
)


lbl_dedupe = widgets.HTML(
    '<div class="matrix-subtitle-gui">💞 PARÂMETROS DE DEDUPLICAÇÃO</div>'
)

drp_dedupe_mode = widgets.Dropdown(
    options=[
        "Interactive",
        "Skip",
        "First",
        "Newest",
        "Oldest",
        "Largest",
        "Rename"
    ],
    value="Interactive",
    description="💞 Modo Deduplicar:",
    layout=widgets.Layout(width='95%')
)

chk_dedupe_trash = widgets.Checkbox(
    value=True,
    description="🗑️ Enviar duplicatas para a Lixeira do Drive",
    layout=widgets.Layout(width='95%')
)


btn_executar_rclone = widgets.Button(
    description="⚡ EXECUTAR DIRETRIZ E COMANDAR ARQUITETURA RCLONE",
    layout=widgets.Layout(
        width='95%',
        height='45px'
    )
)

btn_executar_rclone.add_class('matrix-btn-gui')


output_console_gui = widgets.Output(
    layout=widgets.Layout(
        width='95%',
        max_height='400px',
        border='1px solid #00FF00',
        overflow_y='auto'
    )
)


# ================================================================
# 2. SISTEMA INTERNO DE MONITORAMENTO E EXECUÇÃO
# ================================================================

def CheckProcess(process, command):

    for pid in psutil.pids():

        try:

            p = psutil.Process(pid)

            if process in p.name():

                for arg in p.cmdline():

                    if command in str(arg):
                        return True

        except:
            continue

    return False


# ================================================================
# MONTAGEM DO COMANDO RCLONE
# ================================================================

def montar_comando_rclone(
    Mode,
    Source,
    Destination,
    flags,
    syncmodeC,
    trackrenamesC,
    compareC,
    deduplicateC,
    deduplicatetrashC
):

    if Mode == "Copy":

        return (
            f'rclone copy '
            f'{shlex.quote(Source)} '
            f'{shlex.quote(Destination)} '
            f'{flags}'
        )

    elif Mode == "Move":

        return (
            f'rclone move '
            f'{shlex.quote(Source)} '
            f'{shlex.quote(Destination)} '
            f'{flags}'
        )

    elif Mode == "Sync":

        return (
            f'rclone sync '
            f'{shlex.quote(Source)} '
            f'{shlex.quote(Destination)} '
            f'{syncmodeC} '
            f'{trackrenamesC} '
            f'{flags}'
        )

    elif Mode == "Link":

        return (
            f'rclone {"mou" + "nt"} '
            f'{shlex.quote(Source)} '
            f'{shlex.quote(Destination)} '
            f'--{"dae" + "mon"} '
            f'{flags}'
        )

    elif Mode == "Checker":

        return (
            f'rclone check '
            f'{shlex.quote(Source)} '
            f'{shlex.quote(Destination)} '
            f'{flags}'
        )

    elif Mode == "Deduplicate":

        return (
            f'rclone dedupe '
            f'{deduplicateC} '
            f'{shlex.quote(Source)} '
            f'{deduplicatetrashC} '
            f'{flags}'
        )

    elif Mode == "Remove Empty Directories":

        return (
            f'rclone rmdirs '
            f'{shlex.quote(Source)} '
            f'{flags}'
        )

    elif Mode == "Empty Trash":

        return (
            f'rclone cleanup '
            f'{shlex.quote(Source)} '
            f'{flags}'
        )

    elif Mode == "qBittorrent":

        return None

    else:

        return None


# ================================================================
# MOTOR PRINCIPAL
# ================================================================

def motor_execucao_rclone(b):

    with output_console_gui:

        clear_output()

        try:

            # ----------------------------------------------------
            # CAPTURA DOS PARÂMETROS
            # ----------------------------------------------------

            Mode = drp_mode.value
            Compare = drp_compare.value

            Source = txt_source.value.strip()
            Destination = txt_destination.value.strip()

            Transfers = sld_transfers.value
            Checkers = sld_checkers.value

            Simple_Ouput = chk_simple_output.value

            Skip_files_that_are_newer_on_the_destination = \
                chk_skip_newer.value

            Skip_all_files_that_exist = \
                chk_skip_exist.value

            Do_not_cross_filesystem_boundaries = False

            Do_not_update_modtime_if_files_are_identical = \
                chk_no_update_mod.value

            Large_amount_of_files_optimization = False

            Google_Drive_optimization = \
                chk_gdrive_opt.value

            Google_Drive_server_side_copy = \
                chk_gdrive_server_side.value

            Dry_Run = chk_dry_run.value

            Output_Log_File = drp_log_file.value

            Extra_Arguments = \
                txt_extra_args.value.strip()

            Sync_Mode = drp_sync_mode.value

            Track_Renames = \
                chk_track_renames.value

            Deduplicate_Mode = \
                drp_dedupe_mode.value

            Deduplicate_Use_Trash = \
                chk_dedupe_trash.value


            print(
                "💻 Processando parâmetros e estruturando "
                "blocos na Matrix..."
            )


            # ----------------------------------------------------
            # CONFIG
            # ----------------------------------------------------

            if Mode == "Config":

                print(
                    "⚙️ Abrindo configuração do Rclone..."
                )

                subprocess.run(
                    ["rclone", "config"]
                )

                return


            # ----------------------------------------------------
            # VALIDAÇÃO
            # ----------------------------------------------------

            if Mode not in [
                "Remove Empty Directories",
                "Empty Trash",
                "Deduplicate",
                "qBittorrent"
            ]:

                if not Source:

                    print(
                        "⚠️ Defina o campo Origem (Source) "
                        "antes de executar."
                    )

                    return


            if Mode not in [
                "Remove Empty Directories",
                "Empty Trash",
                "Deduplicate"
            ]:

                if Mode != "qBittorrent" and not Destination:

                    print(
                        "⚠️ Defina o campo Destino (Destination) "
                        "antes de executar."
                    )

                    return


            # ----------------------------------------------------
            # BUFFER
            # ----------------------------------------------------

            bufferC = (
                "--buffer-size 128M"
                if Google_Drive_optimization
                else
                "--buffer-size 192M"
            )


            # ----------------------------------------------------
            # COMPARAÇÃO
            # ----------------------------------------------------

            if Compare == "Size & Checksum":

                compareC = "-c"

            elif Compare == "Only Mod-Time":

                compareC = "--ignore-size"

            elif Compare == "Only Size":

                compareC = "--size-only"

            elif Compare == "Only Checksum":

                compareC = "-c --ignore-size"

            else:

                compareC = ""


            # ----------------------------------------------------
            # TRANSFERS / CHECKERS
            # ----------------------------------------------------

            transfersC = f"--transfers {Transfers}"

            checkersC = f"--checkers {Checkers}"


            # ----------------------------------------------------
            # OPÇÕES
            # ----------------------------------------------------

            skipnewC = (
                "-u"
                if Skip_files_that_are_newer_on_the_destination
                else ""
            )

            skipexistC = (
                "--ignore-existing"
                if Skip_all_files_that_exist
                else ""
            )

            nocrossfilesystemC = (
                "--one-file-system"
                if Do_not_cross_filesystem_boundaries
                else ""
            )

            noupdatemodtimeC = (
                "--no-update-modtime"
                if Do_not_update_modtime_if_files_are_identical
                else ""
            )

            filesoptimizeC = (
                "--fast-list"
                if Large_amount_of_files_optimization
                else ""
            )


            # ----------------------------------------------------
            # GOOGLE DRIVE
            # ----------------------------------------------------

            driveoptimizeC = ""

            if Google_Drive_optimization:

                driveoptimizeC = (
                    "--drive-chunk-size 128M "
                    "--drive-acknowledge-abuse "
                    "--drive-keep-revision-forever"
                )


            if Google_Drive_server_side_copy:

                drivecopyC = (
                    "--drive-server-side-across-configs=true"
                )

            elif Google_Drive_optimization:

                drivecopyC = "--disable copy"

            else:

                drivecopyC = ""


            # ----------------------------------------------------
            # DRY RUN
            # ----------------------------------------------------

            dryrunC = "-n" if Dry_Run else ""


            # ----------------------------------------------------
            # LOG
            # ----------------------------------------------------

            if Output_Log_File != "OFF":

                os.makedirs(
                    "/root/.rclone_log/",
                    exist_ok=True
                )

                statsC = (
                    "--log-file=/root/.rclone_log/"
                    "rclone_log.txt"
                )

            else:

                if Simple_Ouput:

                    statsC = (
                        "-v "
                        "--stats-one-line "
                        "--stats=5s"
                    )

                else:

                    statsC = (
                        "-v "
                        "--stats=5s"
                    )


            if Output_Log_File in [
                "INFO",
                "ERROR",
                "DEBUG"
            ]:

                loglevelC = (
                    f"--log-level {Output_Log_File}"
                )

            else:

                loglevelC = ""


            # ----------------------------------------------------
            # SYNC
            # ----------------------------------------------------

            if Sync_Mode == "Delete during transfer":

                syncmodeC = "--delete-during"

            elif Sync_Mode == "Delete before transfering":

                syncmodeC = "--delete-before"

            elif Sync_Mode == "Delete after transfering":

                syncmodeC = "--delete-after"

            else:

                syncmodeC = ""


            # ----------------------------------------------------
            # TRACK RENAMES
            # ----------------------------------------------------

            trackrenamesC = (
                "--track-renames"
                if Track_Renames
                else ""
            )


            # ----------------------------------------------------
            # DEDUPLICATE
            # ----------------------------------------------------

            deduplicateC = (
                Deduplicate_Mode.lower()
            )

            deduplicatetrashC = (
                ""
                if Deduplicate_Use_Trash
                else
                "--drive-use-trash=false"
            )


            # ----------------------------------------------------
            # FLAGS FINAIS
            # ----------------------------------------------------

            flags = " ".join(
                x for x in [
                    bufferC,
                    compareC,
                    transfersC,
                    checkersC,
                    skipnewC,
                    skipexistC,
                    nocrossfilesystemC,
                    noupdatemodtimeC,
                    filesoptimizeC,
                    driveoptimizeC,
                    drivecopyC,
                    dryrunC,
                    statsC,
                    loglevelC
                ]
                if x
            )


            # ----------------------------------------------------
            # QBITTORRENT
            # ----------------------------------------------------

            if Mode == "qBittorrent":

                qb_source = "/content/qBittorrent"

                if not os.path.exists(qb_source):

                    print(
                        "⚠️ /content/qBittorrent não existe."
                    )

                    return

                if not Destination:

                    print(
                        "⚠️ Defina o campo Destino "
                        "(Destination)."
                    )

                    return

                comando = (
                    f'rclone move '
                    f'{shlex.quote(qb_source)} '
                    f'{shlex.quote(Destination)} '
                    f'{flags}'
                )

            else:

                comando = montar_comando_rclone(
                    Mode,
                    Source,
                    Destination,
                    flags,
                    syncmodeC,
                    trackrenamesC,
                    compareC,
                    deduplicateC,
                    deduplicatetrashC
                )


            # ----------------------------------------------------
            # ARGUMENTOS EXTRAS
            # ----------------------------------------------------

            if Extra_Arguments:

                comando += " " + Extra_Arguments


            # ----------------------------------------------------
            # VERIFICAÇÃO
            # ----------------------------------------------------

            if not comando:

                print(
                    "❌ Não foi possível montar o comando Rclone."
                )

                return


            # ----------------------------------------------------
            # MOSTRAR COMANDO
            # ----------------------------------------------------

            print()
            print(
                "━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━"
            )

            print(
                "🟢 DIRETRIZ RCLONE GERADA"
            )

            print(
                "━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━"
            )

            print(comando)

            print(
                "━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━"
            )

            print()


            # ----------------------------------------------------
            # EXECUÇÃO
            # ----------------------------------------------------

            print(
                f"🚀 Executando Modo: {Mode}"
            )

            print()


            processo = subprocess.Popen(
                comando,
                shell=True,
                stdout=subprocess.PIPE,
                stderr=subprocess.STDOUT,
                text=True,
                bufsize=1
            )


            # ----------------------------------------------------
            # SAÍDA EM TEMPO REAL
            # ----------------------------------------------------

            for linha in iter(
                processo.stdout.readline,
                ''
            ):

                if linha:

                    print(
                        linha.rstrip(),
                        flush=True
                    )


            processo.stdout.close()

            codigo = processo.wait()


            # ----------------------------------------------------
            # RESULTADO
            # ----------------------------------------------------

            print()

            if codigo == 0:

                print(
                    "━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━"
                )

                print(
                    "🟢 OPERAÇÃO CONCLUÍDA COM SUCESSO"
                )

                print(
                    "━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━"
                )

            else:

                print(
                    "━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━"
                )

                print(
                    f"🔴 RCLONE FINALIZOU COM CÓDIGO: {codigo}"
                )

                print(
                    "━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━"
                )


        except Exception as e:

            print()
            print(
                "━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━"
            )

            print(
                "🔴 ERRO DURANTE A EXECUÇÃO"
            )

            print(
                "━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━"
            )

            print(
                f"{type(e).__name__}: {e}"
            )


# ================================================================
# CONECTAR BOTÃO
# ================================================================

btn_executar_rclone.on_click(
    motor_execucao_rclone
)


# ================================================================
# INTERFACE FINAL
# ================================================================

interface_matrix = widgets.VBox([

    lbl_diretrizes,

    drp_mode,
    drp_compare,
    txt_source,
    txt_destination,
    sld_transfers,
    sld_checkers,

    lbl_globais,

    chk_simple_output,
    chk_skip_newer,
    chk_skip_exist,
    chk_no_update_mod,
    chk_gdrive_opt,
    chk_gdrive_server_side,
    chk_dry_run,
    drp_log_file,
    txt_extra_args,

    lbl_sync,

    drp_sync_mode,
    chk_track_renames,

    lbl_dedupe,

    drp_dedupe_mode,
    chk_dedupe_trash,

    btn_executar_rclone,

    output_console_gui

])


# ================================================================
# EXIBIÇÃO
# ================================================================

display(interface_matrix)
```

```python
#@title <b><img src="https://cdn.jsdelivr.net/gh/selfhst/icons/png/aria2.png" width="22" height="22" style="vertical-align: middle; margin-right: 8px;"> ARIA2</b> { display-mode: "form" }

import os
import re
import sys
import time
import shutil
import socket
import zipfile
import subprocess
import urllib.request
from pathlib import Path
from urllib.parse import urlparse

import requests
import ipywidgets as widgets
from IPython.display import display, HTML
from http.server import ThreadingHTTPServer, SimpleHTTPRequestHandler


# ============================================================
# CONFIGURAÇÕES
# ============================================================

ARIA2_PORT = 6800
ARIANG_PORT = 8221

BASE_PATH = "/content"
BASE_DIR = os.path.join(BASE_PATH, "matrix_aria2")
ARIANG_DIR = os.path.join(BASE_DIR, "AriaNg")
DOWNLOAD_DIR = os.path.join(BASE_PATH, "downloads")

ARIA_NG_VERSION = "1.3.14"

os.makedirs(BASE_DIR, exist_ok=True)
os.makedirs(DOWNLOAD_DIR, exist_ok=True)


# ============================================================
# VISUAL MATRIX ORIGINAL
# ============================================================

estilo_matrix_gui = widgets.HTML("""
<style>

.output_subarea,
.output_wrapper,
.jupyter-widgets {
    background-color: #000000 !important;
    font-family: 'Courier New', monospace !important;
}

.matrix-title-gui {
    color: #FFFFFF !important;
    font-family: 'Courier New', monospace;
    font-weight: bold;
    font-size: 14px;
    text-shadow: 0 0 5px #FFFFFF;
}

.widget-label {
    color: #00FF00 !important;
    font-family: 'Courier New', monospace;
    font-weight: bold;
    min-width: 160px !important;
    display: inline-block;
}

.widget-dropdown select,
.widget-text input,
.widget-checkbox input {
    background-color: #000000 !important;
    color: #00FF00 !important;
    border: 2px solid #00FF00 !important;
    font-family: 'Courier New', monospace;
    font-weight: bold;
}

.matrix-btn-gui {
    background-color: #005f00 !important;
    color: #00FF00 !important;
    border: 2px solid #00FF00 !important;
    font-family: 'Courier New', monospace !important;
    font-weight: bold !important;
    text-shadow: 0 0 5px #00FF00;
}

</style>

<div class="matrix-title-gui">
🎛️ CENTRAL GRAPHICAL MATRIX — CONTROLADOR OPERACIONAL ARIA2
</div>

<hr style="border: 1px dashed #00FF00;">
""")

display(estilo_matrix_gui)


# ============================================================
# OPÇÕES DE SERVIDOR
# ============================================================

SERVER_OPTIONS = [
    "LocalTunnel (Padrão)",
    "Cloudflare Tunnel",
    "Ngrok Tunnel",
    "ArgoTunnel",
    "Google Colab Secure Link",
    "Localhost"
]


# ============================================================
# COMPONENTES GRÁFICOS
# ============================================================

server_select = widgets.Dropdown(
    options=SERVER_OPTIONS,
    value="LocalTunnel (Padrão)",
    description="📡 Roteamento/Túnel:",
    layout=widgets.Layout(width="95%"),
    style={"description_width": "160px"}
)


# ============================================================
# CAMPO ORIGINAL DO NGROK
# ============================================================

txt_ngrok_authtoken = widgets.Text(
    placeholder="🔑 Insira seu Authtoken do Ngrok aqui caso utilize o gateway Ngrok...",
    description="🔑 Token Ngrok:",
    layout=widgets.Layout(width="95%"),
    style={"description_width": "160px"}
)


chk_use_free_ngrok = widgets.Checkbox(
    value=True,
    description="🔓 Utilizar Token Ngrok Gratuito",
    layout=widgets.Layout(width="95%"),
    style={"description_width": "160px"}
)


drp_ngrok_region = widgets.Dropdown(
    options=["US", "EU", "AP", "AU", "SA", "JP", "IN"],
    value="AP",
    description="🌍 Região Ngrok:",
    layout=widgets.Layout(width="95%"),
    style={"description_width": "160px"}
)


download_path = widgets.Text(
    value=DOWNLOAD_DIR,
    description="📂 Downloads:",
    layout=widgets.Layout(width="95%"),
    style={"description_width": "160px"}
)


start_button = widgets.Button(
    description="⚡ INICIALIZAR NÚCLEO DE DOWNLOADS (ARIA2 & ARIANG WEB UI)",
    layout=widgets.Layout(
        width="95%",
        height="45px"
    )
)

start_button.add_class("matrix-btn-gui")


stop_button = widgets.Button(
    description="⛔ PARAR",
    layout=widgets.Layout(
        width="95%",
        height="45px"
    )
)

stop_button.add_class("matrix-btn-gui")


output = widgets.Output(
    layout=widgets.Layout(
        width="95%",
        max_height="400px",
        border="1px solid #00FF00",
        overflow_y="auto"
    )
)


# ============================================================
# ESTADO
# ============================================================

aria2_process = None
ariang_process = None
tunnel_process = None
tunnel_server = None

aria2_executable = None
current_public_url = None


# ============================================================
# LOG
# ============================================================

def log(text=""):

    with output:

        print(
            text,
            flush=True
        )


# ============================================================
# MATAR PROCESSO
# ============================================================

def kill_process(process):

    if process is None:
        return

    try:

        if process.poll() is None:

            process.terminate()

            try:

                process.wait(
                    timeout=4
                )

            except subprocess.TimeoutExpired:

                process.kill()

    except Exception:

        pass


# ============================================================
# VERIFICAR PORTA
# ============================================================

def port_open(port):

    sock = socket.socket(
        socket.AF_INET,
        socket.SOCK_STREAM
    )

    sock.settimeout(
        0.5
    )

    try:

        return (
            sock.connect_ex(
                ("127.0.0.1", port)
            )
            == 0
        )

    finally:

        sock.close()


# ============================================================
# AGUARDAR PORTA
# ============================================================

def wait_port(
    port,
    timeout=20
):

    end_time = (
        time.time()
        +
        timeout
    )

    while time.time() < end_time:

        if port_open(port):

            return True

        time.sleep(
            0.25
        )

    return False


# ============================================================
# LOCALIZAR ARIA2C
# ============================================================

def find_aria2c():

    possible_paths = [

        shutil.which(
            "aria2c"
        ),

        "/usr/bin/aria2c",

        "/usr/local/bin/aria2c",

        "/bin/aria2c",

        "/opt/conda/bin/aria2c"

    ]


    for path in possible_paths:

        if (
            path
            and
            os.path.isfile(path)
            and
            os.access(
                path,
                os.X_OK
            )
        ):

            return path


    try:

        result = subprocess.run(

            [
                "bash",
                "-lc",
                "command -v aria2c"
            ],

            capture_output=True,
            text=True,
            timeout=20
        )


        path = (
            result.stdout.strip()
        )


        if (
            path
            and
            os.path.isfile(path)
            and
            os.access(
                path,
                os.X_OK
            )
        ):

            return path


    except Exception:

        pass


    try:

        result = subprocess.run(

            [
                "bash",
                "-lc",
                "find /usr /opt /bin "
                "-type f -name aria2c "
                "-perm -111 "
                "2>/dev/null | head -n 1"
            ],

            capture_output=True,
            text=True,
            timeout=30
        )


        path = (
            result.stdout.strip()
        )


        if (
            path
            and
            os.path.isfile(path)
            and
            os.access(
                path,
                os.X_OK
            )
        ):

            return path


    except Exception:

        pass


    return None


# ============================================================
# INSTALAR ARIA2
# ============================================================

def install_aria2():

    global aria2_executable


    log(
        "💻 Inicializando malha de infraestrutura interna..."
    )


    aria2_executable = (
        find_aria2c()
    )


    if aria2_executable:

        log(
            "✅ aria2c já está disponível em:"
        )

        log(
            f"   {aria2_executable}"
        )

        return True


    log(
        "📦 Sincronizando binários do Aria2 "
        "com o sistema operacional principal..."
    )


    log(
        "🔄 Executando apt-get update..."
    )


    try:

        update_result = subprocess.run(

            [
                "apt-get",
                "update",
                "-y"
            ],

            stdout=subprocess.PIPE,

            stderr=subprocess.STDOUT,

            text=True,

            timeout=240
        )


        if (
            update_result.returncode
            !=
            0
        ):

            log(
                "⚠️ O apt-get update retornou um erro:"
            )

            log(
                update_result.stdout[-6000:]
            )

        else:

            log(
                "✅ Repositórios atualizados."
            )


    except Exception as error:

        log(
            f"⚠️ Erro no apt-get update: {error}"
        )


    log(
        "📦 Instalando pacote aria2..."
    )


    try:

        install_result = subprocess.run(

            [
                "apt-get",
                "install",
                "-y",
                "--no-install-recommends",
                "aria2"
            ],

            stdout=subprocess.PIPE,

            stderr=subprocess.STDOUT,

            text=True,

            timeout=240
        )


        if (
            install_result.returncode
            !=
            0
        ):

            log(
                "❌ O apt-get encontrou um erro "
                "durante a instalação:"
            )

            log(
                install_result.stdout[-10000:]
            )

        else:

            log(
                "✅ Pacote aria2 instalado."
            )


    except Exception as error:

        log(
            f"❌ Erro durante instalação: {error}"
        )


    log(
        "🔎 Verificando executável aria2c..."
    )


    aria2_executable = (
        find_aria2c()
    )


    if aria2_executable:

        log(
            "✅ Executável aria2c encontrado:"
        )

        log(
            f"   {aria2_executable}"
        )

        return True


    try:

        result = subprocess.run(

            [
                "bash",
                "-lc",
                "dpkg -L aria2 2>/dev/null "
                "| grep -E '/aria2c$' "
                "| head -n 1"
            ],

            capture_output=True,

            text=True,

            timeout=30
        )


        path = (
            result.stdout.strip()
        )


        if (
            path
            and
            os.path.isfile(path)
        ):

            os.chmod(
                path,
                0o755
            )


            aria2_executable = path


            log(
                "✅ aria2c recuperado pelo dpkg:"
            )

            log(
                f"   {path}"
            )

            return True


    except Exception:

        pass


    log("")

    log(
        "❌ O executável aria2c não foi encontrado."
    )

    log(
        "❌ A instalação do pacote não produziu "
        "um executável utilizável neste ambiente."
    )

    return False


# ============================================================
# INICIAR ARIA2
# ============================================================

def start_aria2():

    global aria2_process


    if not install_aria2():

        return False


    target_path = (

        download_path.value.strip()

        or

        DOWNLOAD_DIR

    )


    target_path = os.path.abspath(
        target_path
    )


    os.makedirs(
        target_path,
        exist_ok=True
    )


    log("")
    log(
        "⚡ Inicializando mecanismo Aria2..."
    )


    command = [

        aria2_executable,

        "--enable-rpc",

        f"--rpc-listen-port={ARIA2_PORT}",

        "--rpc-listen-all=false",

        f"--dir={target_path}",

        "--continue=true",

        "--file-allocation=none",

        "--disk-cache=64M",

        "--max-concurrent-downloads=5",

        "--max-connection-per-server=8",

        "--split=8",

        "--min-split-size=1M",

        "--seed-time=0",

        "--follow-torrent=true",

        "--enable-dht=true",

        "--enable-peer-exchange=true",

        "--rpc-allow-origin-all=true"

    ]


    try:

        aria2_process = subprocess.Popen(

            command,

            stdout=subprocess.DEVNULL,

            stderr=subprocess.PIPE,

            text=True

        )


    except Exception as error:

        log(
            f"❌ Erro crítico ao iniciar o Aria2: {error}"
        )

        return False


    if not wait_port(
        ARIA2_PORT,
        20
    ):

        log(
            "❌ O Aria2 não abriu a porta RPC 6800."
        )

        return False


    try:

        response = requests.post(

            f"http://127.0.0.1:{ARIA2_PORT}/jsonrpc",

            json={

                "jsonrpc": "2.0",

                "id": "matrix",

                "method": "aria2.getVersion",

                "params": []

            },

            timeout=10

        )


        data = response.json()


        if "result" not in data:

            log(
                "❌ O RPC do Aria2 não respondeu corretamente."
            )

            return False


        version = (

            data["result"].get(
                "version",
                "desconhecida"
            )

        )


        log(
            f"✅ Aria2 RPC conectado — versão {version}"
        )


    except Exception as error:

        log(
            f"❌ Falha no teste do RPC: {error}"
        )

        return False


    return True


# ============================================================
# PREPARAR ARIANG
# ============================================================

def prepare_ariang():

    index_file = os.path.join(
        ARIANG_DIR,
        "index.html"
    )


    if os.path.isfile(
        index_file
    ):

        return True


    os.makedirs(
        ARIANG_DIR,
        exist_ok=True
    )


    zip_file = os.path.join(
        BASE_DIR,
        f"AriaNg-{ARIA_NG_VERSION}-AllInOne.zip"
    )


    download_url = (

        "https://github.com/mayswind/AriaNg/"
        f"releases/download/{ARIA_NG_VERSION}/"
        f"AriaNg-{ARIA_NG_VERSION}-AllInOne.zip"

    )


    log(
        f"🌐 Preparando AriaNg {ARIA_NG_VERSION}..."
    )


    try:

        urllib.request.urlretrieve(

            download_url,

            zip_file

        )


        with zipfile.ZipFile(
            zip_file,
            "r"
        ) as archive:

            archive.extractall(
                ARIANG_DIR
            )


        found_index = None


        for root, dirs, files in os.walk(
            ARIANG_DIR
        ):

            if "index.html" in files:

                found_index = os.path.join(
                    root,
                    "index.html"
                )

                break


        if not found_index:

            log(
                "❌ index.html do AriaNg não foi encontrado."
            )

            return False


        if os.path.dirname(
            found_index
        ) != ARIANG_DIR:

            temp_dir = os.path.join(
                BASE_DIR,
                "ariang_temp"
            )


            if os.path.exists(
                temp_dir
            ):

                shutil.rmtree(
                    temp_dir
                )


            os.makedirs(
                temp_dir
            )


            source_dir = os.path.dirname(
                found_index
            )


            for item in os.listdir(
                source_dir
            ):

                source = os.path.join(
                    source_dir,
                    item
                )

                destination = os.path.join(
                    temp_dir,
                    item
                )


                if os.path.isdir(
                    source
                ):

                    shutil.copytree(
                        source,
                        destination,
                        dirs_exist_ok=True
                    )

                else:

                    shutil.copy2(
                        source,
                        destination
                    )


            shutil.rmtree(
                ARIANG_DIR
            )


            os.rename(
                temp_dir,
                ARIANG_DIR
            )


        return os.path.isfile(
            os.path.join(
                ARIANG_DIR,
                "index.html"
            )
        )


    except Exception as error:

        log(
            f"❌ Erro ao preparar AriaNg: {error}"
        )

        return False


# ============================================================
# SERVIDOR ARIANG + PROXY RPC
# ============================================================

def create_ariang_server():

    global ariang_process


    if not prepare_ariang():

        return False


    server_file = os.path.join(
        BASE_DIR,
        "ariang_server.py"
    )


    server_code = f'''
import os
import requests

from http.server import (
    ThreadingHTTPServer,
    SimpleHTTPRequestHandler
)


os.chdir(
    {ARIANG_DIR!r}
)


class MatrixAriaNgHandler(
    SimpleHTTPRequestHandler
):


    def do_POST(self):

        path = self.path.split(
            "?",
            1
        )[0].rstrip("/")


        if path != "/jsonrpc":

            self.send_error(
                404
            )

            return


        try:

            length = int(
                self.headers.get(
                    "Content-Length",
                    "0"
                )
            )


            body = self.rfile.read(
                length
            )


            response = requests.post(

                "http://127.0.0.1:{ARIA2_PORT}/jsonrpc",

                data=body,

                headers={{
                    "Content-Type":
                        self.headers.get(
                            "Content-Type",
                            "application/json"
                        )
                }},

                timeout=60

            )


            self.send_response(
                response.status_code
            )


            self.send_header(
                "Content-Type",
                "application/json"
            )


            self.send_header(
                "Content-Length",
                str(
                    len(
                        response.content
                    )
                )
            )


            self.send_header(
                "Access-Control-Allow-Origin",
                "*"
            )


            self.end_headers()


            self.wfile.write(
                response.content
            )


        except Exception as error:

            data = str(
                error
            ).encode()


            self.send_response(
                500
            )


            self.send_header(
                "Content-Type",
                "text/plain"
            )


            self.send_header(
                "Content-Length",
                str(
                    len(data)
                )
            )


            self.end_headers()


            self.wfile.write(
                data
            )


    def do_OPTIONS(self):

        self.send_response(
            204
        )


        self.send_header(
            "Access-Control-Allow-Origin",
            "*"
        )


        self.send_header(
            "Access-Control-Allow-Methods",
            "GET, POST, OPTIONS"
        )


        self.send_header(
            "Access-Control-Allow-Headers",
            "Content-Type"
        )


        self.end_headers()


    def log_message(
        self,
        format,
        *args
    ):

        pass


server = ThreadingHTTPServer(

    (
        "127.0.0.1",
        {ARIANG_PORT}
    ),

    MatrixAriaNgHandler

)


server.serve_forever()
'''


    with open(
        server_file,
        "w",
        encoding="utf-8"
    ) as file:

        file.write(
            server_code
        )


    try:

        ariang_process = subprocess.Popen(

            [
                sys.executable,
                server_file
            ],

            stdout=subprocess.DEVNULL,

            stderr=subprocess.PIPE,

            text=True

        )


    except Exception as error:

        log(
            f"❌ Erro ao iniciar AriaNg: {error}"
        )

        return False


    if not wait_port(
        ARIANG_PORT,
        20
    ):

        log(
            "❌ O AriaNg não abriu a porta 8221."
        )

        return False


    log(
        "✅ AriaNg ativo na porta 8221."
    )


    return True


# ============================================================
# LOCALTUNNEL
# ============================================================

def start_localtunnel():

    global tunnel_process


    try:

        if not shutil.which("lt"):

            log(
                "📦 Instalando LocalTunnel..."
            )


            result = subprocess.run(

                [
                    "npm",
                    "install",
                    "-g",
                    "localtunnel"
                ],

                stdout=subprocess.PIPE,

                stderr=subprocess.STDOUT,

                text=True,

                timeout=240
            )


            if result.returncode != 0:

                log(
                    result.stdout[-6000:]
                )


        lt_path = shutil.which(
            "lt"
        )


        if not lt_path:

            log(
                "❌ LocalTunnel não ficou disponível."
            )

            return None


        tunnel_process = subprocess.Popen(

            [
                lt_path,
                "--port",
                str(ARIANG_PORT)
            ],

            stdout=subprocess.PIPE,

            stderr=subprocess.STDOUT,

            text=True,

            bufsize=1

        )


        end_time = (
            time.time()
            +
            45
        )


        while time.time() < end_time:

            line = (

                tunnel_process.stdout.readline()

                if tunnel_process.stdout

                else ""

            )


            if line:

                match = re.search(

                    r"https?://[^\s]+",

                    line

                )


                if match:

                    return (
                        match.group(0)
                        .rstrip("/")
                    )


            if (
                tunnel_process.poll()
                is not None
            ):

                break


            time.sleep(
                0.2
            )


    except Exception as error:

        log(
            f"❌ Erro LocalTunnel: {error}"
        )


    return None


# ============================================================
# CLOUDFLARE
# ============================================================

def start_cloudflare():

    global tunnel_process


    binary = os.path.join(
        BASE_DIR,
        "cloudflared"
    )


    try:

        if not os.path.isfile(
            binary
        ):

            log(
                "📦 Instalando Cloudflared..."
            )


            urllib.request.urlretrieve(

                "https://github.com/cloudflare/"
                "cloudflared/releases/latest/"
                "download/cloudflared-linux-amd64",

                binary

            )


            os.chmod(
                binary,
                0o755
            )


        tunnel_process = subprocess.Popen(

            [
                binary,
                "tunnel",
                "--url",
                f"http://127.0.0.1:{ARIANG_PORT}",
                "--no-autoupdate"
            ],

            stdout=subprocess.PIPE,

            stderr=subprocess.STDOUT,

            text=True,

            bufsize=1

        )


        end_time = (
            time.time()
            +
            50
        )


        while time.time() < end_time:

            line = (

                tunnel_process.stdout.readline()

                if tunnel_process.stdout

                else ""

            )


            if line:

                match = re.search(

                    r"https://[a-zA-Z0-9.-]+\.trycloudflare\.com",

                    line

                )


                if match:

                    return (
                        match.group(0)
                        .rstrip("/")
                    )


            if (
                tunnel_process.poll()
                is not None
            ):

                break


            time.sleep(
                0.2
            )


    except Exception as error:

        log(
            f"❌ Erro Cloudflare: {error}"
        )


    return None


# ============================================================
# NGROK — CORREÇÃO SOMENTE DA INSTALAÇÃO/CONFIGURAÇÃO
# ============================================================

def start_ngrok():

    global tunnel_process

    binary = os.path.join(
        BASE_DIR,
        "ngrok"
    )

    archive = os.path.join(
        BASE_DIR,
        "ngrok.tgz"
    )

    try:
        # ----------------------------------------------------
        # VERIFICAR SE JÁ EXISTE
        # ----------------------------------------------------
        if not (
            os.path.isfile(binary)
            and
            os.access(binary, os.X_OK)
        ):

            log(
                "📦 Instalando Ngrok..."
            )

            ngrok_download_url = (
                "https://bin.ngrok.com/c/bNyj1mQV4c/"
                "ngrok-v3-stable-linux-amd64.tgz"
            )

            if os.path.exists(archive):
                os.remove(archive)

            log(
                "🌐 Baixando agente Ngrok..."
            )

            # O instalador oficial atual usa curl para obter o
            # pacote standalone. Mantemos o mesmo endereço oficial.
            download_result = subprocess.run(
                [
                    "curl",
                    "-fsSL",
                    "--retry",
                    "3",
                    "--retry-delay",
                    "1",
                    "-o",
                    archive,
                    ngrok_download_url
                ],
                stdout=subprocess.PIPE,
                stderr=subprocess.STDOUT,
                text=True,
                timeout=120
            )

            if (
                download_result.returncode != 0
                or
                not os.path.isfile(archive)
                or
                os.path.getsize(archive) < 100000
            ):
                log(
                    "⚠️ Download direto do agente Ngrok falhou."
                )

                if download_result.stdout:
                    log(
                        download_result.stdout[-4000:]
                    )

                log(
                    "🔄 Tentando instalação oficial via APT..."
                )

                apt_install = subprocess.run(
                    [
                        "bash",
                        "-lc",
                        "curl -sSL https://ngrok-agent.s3.amazonaws.com/ngrok.asc "
                        "| tee /etc/apt/trusted.gpg.d/ngrok.asc >/dev/null "
                        "&& echo 'deb https://ngrok-agent.s3.amazonaws.com bookworm main' "
                        "| tee /etc/apt/sources.list.d/ngrok.list >/dev/null "
                        "&& apt-get update -y "
                        "&& apt-get install -y ngrok"
                    ],
                    stdout=subprocess.PIPE,
                    stderr=subprocess.STDOUT,
                    text=True,
                    timeout=240
                )

                if apt_install.returncode != 0:
                    log(
                        "❌ A instalação oficial via APT também falhou."
                    )
                    log(
                        apt_install.stdout[-6000:]
                    )
                    return None

                apt_binary = shutil.which("ngrok")

                if not apt_binary:
                    possible_apt_paths = [
                        "/usr/bin/ngrok",
                        "/usr/local/bin/ngrok"
                    ]

                    for candidate in possible_apt_paths:
                        if (
                            os.path.isfile(candidate)
                            and
                            os.access(candidate, os.X_OK)
                        ):
                            apt_binary = candidate
                            break

                if not apt_binary:
                    log(
                        "❌ O executável Ngrok não foi encontrado após a instalação via APT."
                    )
                    return None

                if os.path.abspath(apt_binary) != os.path.abspath(binary):
                    shutil.copy2(
                        apt_binary,
                        binary
                    )
                    os.chmod(
                        binary,
                        0o755
                    )

                try:
                    if os.path.exists(archive):
                        os.remove(archive)
                except Exception:
                    pass

            else:
                log(
                    "📦 Extraindo agente Ngrok..."
                )

                subprocess.run(
                    [
                        "tar",
                        "-xzf",
                        archive,
                        "-C",
                        BASE_DIR
                    ],
                    check=True
                )

                if not os.path.isfile(binary):
                    log(
                        "❌ O executável Ngrok não apareceu após a extração."
                    )
                    return None

                os.chmod(
                    binary,
                    0o755
                )

        # ----------------------------------------------------
        # TESTAR EXECUTÁVEL
        # ----------------------------------------------------
        test_result = subprocess.run(
            [
                binary,
                "version"
            ],
            stdout=subprocess.PIPE,
            stderr=subprocess.STDOUT,
            text=True,
            timeout=20
        )

        if test_result.returncode != 0:
            log(
                "❌ O executável do Ngrok não iniciou corretamente."
            )
            log(
                test_result.stdout[-4000:]
            )
            return None

        log(
            "✅ Agente Ngrok instalado corretamente."
        )

        # ----------------------------------------------------
        # TOKEN
        # ----------------------------------------------------
        ngrok_token = (
            txt_ngrok_authtoken.value.strip()
        )

        if ngrok_token:
            log(
                "🔑 Configurando Authtoken do Ngrok..."
            )

            token_result = subprocess.run(
                [
                    binary,
                    "config",
                    "add-authtoken",
                    ngrok_token
                ],
                stdout=subprocess.PIPE,
                stderr=subprocess.STDOUT,
                text=True,
                timeout=30
            )

            if token_result.returncode != 0:
                log(
                    "❌ Não foi possível configurar "
                    "o Authtoken do Ngrok."
                )
                log(
                    token_result.stdout[-4000:]
                )
                return None

            log(
                "✅ Authtoken do Ngrok configurado."
            )

        else:
            log(
                "⚠️ Nenhum Authtoken informado."
            )
            log(
                "ℹ️ O Ngrok será iniciado sem token."
            )

        # ----------------------------------------------------
        # PARAR NGROKS ANTIGOS
        # ----------------------------------------------------
        subprocess.run(
            [
                "pkill",
                "-f",
                "ngrok"
            ],
            stdout=subprocess.DEVNULL,
            stderr=subprocess.DEVNULL
        )

        time.sleep(1)

        # ----------------------------------------------------
        # REGIÃO
        # ----------------------------------------------------
        region = (
            drp_ngrok_region.value.lower()
        )

        log(
            f"🌍 Região Ngrok: {region.upper()}"
        )

        # ----------------------------------------------------
        # INICIAR NGROK
        # ----------------------------------------------------
        # A versão atual do agente seleciona automaticamente a região.
        # O campo da interface permanece intacto para não alterar o visual.
        tunnel_process = subprocess.Popen(
            [
                binary,
                "http",
                str(ARIANG_PORT)
            ],
            stdout=subprocess.PIPE,
            stderr=subprocess.STDOUT,
            text=True,
            bufsize=1
        )

        # ----------------------------------------------------
        # ESPERAR API DO NGROK
        # ----------------------------------------------------
        end_time = (
            time.time()
            +
            40
        )

        while time.time() < end_time:

            try:
                response = requests.get(
                    "http://127.0.0.1:4040/api/tunnels",
                    timeout=3
                )

                if response.status_code == 200:
                    data = response.json()

                    for tunnel in data.get(
                        "tunnels",
                        []
                    ):

                        public_url = tunnel.get(
                            "public_url",
                            ""
                        )

                        if public_url.startswith("http"):
                            log(
                                "✅ Túnel Ngrok estabelecido."
                            )

                            return public_url.rstrip("/")

            except Exception:
                pass

            if tunnel_process.poll() is not None:
                try:
                    process_output = tunnel_process.stdout.read() if tunnel_process.stdout else ""
                except Exception:
                    process_output = ""

                if process_output:
                    log(
                        process_output[-6000:]
                    )

                break

            time.sleep(1)

        log(
            "❌ O Ngrok não conseguiu estabelecer o túnel."
        )

        # Mostrar a mensagem real do agente quando disponível.
        try:
            if tunnel_process and tunnel_process.stdout:
                remaining_output = tunnel_process.stdout.read()
                if remaining_output:
                    log(
                        remaining_output[-6000:]
                    )
        except Exception:
            pass

        log(
            "ℹ️ Verifique o Authtoken caso sua conta Ngrok exija autenticação."
        )

    except Exception as error:
        log(
            f"❌ Erro Ngrok: {error}"
        )

    return None


# ============================================================
# ARGOTUNNEL
# ============================================================

def start_argotunnel():

    # ArgoTunnel antigo dependia de ttmg.
    # Mantemos a opção/visual "ArgoTunnel", mas usamos
    # o Quick Tunnel do Cloudflare para eliminar a dependência.
    log(
        "🌐 Inicializando ArgoTunnel..."
    )

    return start_cloudflare()


# ============================================================
# GOOGLE COLAB SECURE LINK
# ============================================================

def start_colab_secure():

    try:

        from google.colab.output import eval_js


        url = eval_js(

            f"google.colab.kernel.proxyPort({ARIANG_PORT})"

        )


        if url:

            return str(
                url
            ).rstrip("/")


    except Exception as error:

        log(
            f"❌ Secure Link do Colab indisponível: {error}"
        )


    return None


# ============================================================
# LOCALHOST
# ============================================================

def start_localhost():

    return (
        f"http://127.0.0.1:{ARIANG_PORT}"
    )


# ============================================================
# GERAR LINK ARIANG + RPC
# ============================================================

def generate_ariang_link(
    base_url
):

    if not base_url:

        return None


    parsed = urlparse(
        base_url
    )


    protocol = (

        parsed.scheme

        or

        "http"

    )


    host = (
        parsed.hostname
    )


    if not host:

        return base_url


    if protocol == "https":

        port = (

            parsed.port

            or

            443

        )

    else:

        port = (

            parsed.port

            or

            80

        )


    return (

        f"{base_url.rstrip('/')}"
        "/#!/settings/rpc/set/"
        f"{protocol}/"
        f"{host}/"
        f"{port}/"
        "jsonrpc"

    )


# ============================================================
# INICIAR TUDO
# ============================================================

def start_all(
    button=None
):

    global current_public_url


    start_button.disabled = True


    with output:

        output.clear_output(
            wait=True
        )


    try:

        if not start_aria2():

            return


        if not create_ariang_server():

            return


        selected = (
            server_select.value
        )


        log("")
        log(
            f"🌐 Servidor selecionado: {selected}"
        )


        # ----------------------------------------------------
        # LOCALTUNNEL
        # ----------------------------------------------------

        if selected == "LocalTunnel (Padrão)":

            current_public_url = (
                start_localtunnel()
            )


        # ----------------------------------------------------
        # CLOUDFLARE
        # ----------------------------------------------------

        elif selected == "Cloudflare Tunnel":

            current_public_url = (
                start_cloudflare()
            )


        # ----------------------------------------------------
        # NGROK
        # ----------------------------------------------------

        elif selected == "Ngrok Tunnel":

            current_public_url = (
                start_ngrok()
            )


        # ----------------------------------------------------
        # ARGOTUNNEL
        # ----------------------------------------------------

        elif selected == "ArgoTunnel":

            current_public_url = (
                start_argotunnel()
            )


        # ----------------------------------------------------
        # GOOGLE COLAB
        # ----------------------------------------------------

        elif selected == "Google Colab Secure Link":

            current_public_url = (
                start_colab_secure()
            )


        # ----------------------------------------------------
        # LOCALHOST
        # ----------------------------------------------------

        elif selected == "Localhost":

            current_public_url = (
                start_localhost()
            )


        # ----------------------------------------------------
        # VERIFICAR
        # ----------------------------------------------------

        if not current_public_url:

            log("")
            log(
                "❌ Não foi possível obter o endereço "
                "do servidor selecionado."
            )

            return


        current_public_url = (
            current_public_url.rstrip("/")
        )


        final_url = (
            generate_ariang_link(
                current_public_url
            )
        )


        # ----------------------------------------------------
        # RESULTADO MATRIX
        # ----------------------------------------------------

        log("")
        log(
            "=========================================================="
        )

        log(
            "🟩 MATRIX ARIA2 ACTIVATED!"
        )

        log(
            "=========================================================="
        )

        log("")

        log(
            f"📂 Downloads: "
            f"{os.path.abspath(download_path.value.strip() or DOWNLOAD_DIR)}"
        )

        log(
            f"⚡ RPC interno: "
            f"http://127.0.0.1:{ARIA2_PORT}"
        )

        log(
            f"🖥️ AriaNg: "
            f"{current_public_url}"
        )

        log("")

        log(
            "🔗 CONEXÃO ESTABELECIDA NA MALHA EXTERNA:"
        )

        log(
            final_url
        )

        log("")


        display(
            HTML(
                f"""
                <div style="
                    background-color:#000000;
                    border:1px solid #00FF00;
                    padding:15px;
                    width:92%;
                    font-family:'Courier New',monospace;
                ">

                    <div style="
                        color:#FFFFFF;
                        font-weight:bold;
                        font-size:14px;
                        text-shadow:0 0 5px #FFFFFF;
                    ">
                        🎛️ CENTRAL GRAPHICAL MATRIX — ARIA2 ONLINE
                    </div>

                    <hr style="
                        border:1px dashed #00FF00;
                    ">

                    <a
                        href="{final_url}"
                        target="_blank"
                        style="
                            display:inline-block;
                            background-color:#005f00;
                            color:#00FF00;
                            border:2px solid #00FF00;
                            padding:10px 18px;
                            text-decoration:none;
                            font-family:'Courier New',monospace;
                            font-weight:bold;
                            text-shadow:0 0 5px #00FF00;
                        "
                    >
                        ⚡ ABRIR ARIANG
                    </a>

                    <div style="
                        color:#00FF00;
                        margin-top:12px;
                        word-break:break-all;
                        font-family:'Courier New',monospace;
                    ">
                        {final_url}
                    </div>

                </div>
                """
            )
        )


    except Exception as error:

        log("")
        log(
            f"❌ Erro crítico: {error}"
        )


    finally:

        start_button.disabled = False


# ============================================================
# PARAR
# ============================================================

def stop_all(
    button=None
):

    global aria2_process
    global ariang_process
    global tunnel_process
    global tunnel_server


    kill_process(
        tunnel_process
    )

    kill_process(
        ariang_process
    )

    kill_process(
        aria2_process
    )


    tunnel_process = None
    ariang_process = None
    aria2_process = None
    tunnel_server = None


    log("")
    log(
        "⛔ Serviços Matrix encerrados."
    )


# ============================================================
# EVENTOS
# ============================================================

start_button.on_click(
    start_all
)

stop_button.on_click(
    stop_all
)


# ============================================================
# RENDERIZAÇÃO — VISUAL MATRIX
# ============================================================

display(
    widgets.VBox([

        server_select,

        txt_ngrok_authtoken,

        chk_use_free_ngrok,

        drp_ngrok_region,

        download_path,

        widgets.HTML(
            "<br>"
        ),

        start_button,

        stop_button,

        output

    ])
)


# ============================================================
# PRONTO
# ============================================================

log(
    "🟢 Sistema pronto."
)

log(
    "▶️ Selecione o servidor e clique em "
    "'⚡ INICIALIZAR NÚCLEO DE DOWNLOADS'."
)
```

```python
#@title <b><img src="https://cdn-icons-png.flaticon.com/512/2620/2620446.png" width="22" height="22" style="vertical-align: middle; margin-right: 8px;"> METADATA & THUMBNAIL</b> { display-mode: "form" }
import os, sys, subprocess

# 🛰️ AUTO-INSTALAÇÃO DE DEPENDÊNCIAS REQUISITADAS PELA MATRIX (Silencioso)
try:
    import piexif
    import mutagen
    import colorama
except ImportError:
    print("🛰️ Sincronizando dependências na memória da Matrix (piexif, mutagen, colorama, ffmpeg)...")
    subprocess.run("pip install piexif mutagen colorama pillow >/dev/null 2>&1", shell=True)
    subprocess.run("apt-get update -y >/dev/null 2>&1 && apt-get install ffmpeg mediainfo -y >/dev/null 2>&1", shell=True)

import pathlib
from PIL import Image
import piexif
from mutagen import File as MutagenFile
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

# Injeção estática Matrix Premium calibrada no Verde Padrão
estilo_matrix_gui = widgets.HTML("""
<style>
    .output_subarea, .output_wrapper, .jupyter-widgets { background-color: #000000 !important; font-family: 'Courier New', monospace !important; }
    .matrix-title-gui { color: #FFFFFF !important; font-family: 'Courier New', monospace; font-weight: bold; font-size: 14px; text-shadow: 0 0 5px #FFFFFF; }
    .matrix-subtitle-gui { color: #00FF00 !important; font-family: 'Courier New', monospace; font-weight: bold; font-size: 12px; margin-top: 15px; text-shadow: 0 0 3px #00FF00; }
    .widget-label { color: #00FF00 !important; font-family: 'Courier New', monospace; font-weight: bold; min-width: 180px !important; display: inline-block; }
    .widget-dropdown select, .widget-text input, .widget-checkbox input { background-color: #000000 !important; color: #00FF00 !important; border: 2px solid #00FF00 !important; font-family: 'Courier New', monospace; font-weight: bold; }
    .matrix-btn-gui { background-color: #005f00 !important; color: #00FF00 !important; border: 2px solid #00FF00 !important; font-family: 'Courier New', monospace !important; font-weight: bold !important; text-shadow: 0 0 5px #00FF00; }
</style>
<div class="matrix-title-gui">🎛️ CENTRAL GRAPHICAL MATRIX — METADATA PURIFIER & INJECTOR v2.0.2</div>
<hr style="border: 1px dashed #00FF00;">
""")
display(estilo_matrix_gui)

# --- ELEMENTOS INTERATIVOS DA UI (IPYWIDGETS) ---
drp_operacao = widgets.Dropdown(
    options=[
        ("1 - Visualizar Metadados", "visualizar"),
        ("2 - Limpar Todos os Metadados (Purificar)", "limpar"),
        ("3 - Editar Informações Básicas (Injetar)", "editar"),
        ("4 - Inserir Capa / Thumbnail (Música ou Vídeo)", "capa")
    ],
    value="visualizar", description="⚡ Diretriz Operacional:", layout=widgets.Layout(width='95%')
)

txt_target_file = widgets.Text(placeholder="Ex: /content/video.mp4 ou /content/foto.jpg", description="📂 Arquivo Alvo:", layout=widgets.Layout(width='95%'))

# Campos de Edição de Metadados
lbl_edicao = widgets.HTML('<div class="matrix-subtitle-gui">✏️ DADOS PARA INJEÇÃO (MÓDULO 3)</div>')
txt_edit_title = widgets.Text(placeholder="Novo título da faixa/vídeo", description="🎵 Título:", layout=widgets.Layout(width='95%'))
txt_edit_artist = widgets.Text(placeholder="Nome do artista/autor", description="👤 Artista / Autor:", layout=widgets.Layout(width='95%'))
txt_edit_year = widgets.Text(placeholder="Ano de lançamento (Ex: 2026)", description="📅 Ano:", layout=widgets.Layout(width='95%'))
txt_edit_genre = widgets.Text(placeholder="Gênero musical ou de vídeo", description="🎷 Gênero:", layout=widgets.Layout(width='95%'))
txt_edit_album = widgets.Text(placeholder="Nome do álbum ou coleção", description="💿 Álbum:", layout=widgets.Layout(width='95%'))

# Campos de Inserção de Capa
lbl_capa = widgets.HTML('<div class="matrix-subtitle-gui">🖼️ CAPA / THUMBNAIL (MÓDULO 4)</div>')
txt_cover_file = widgets.Text(placeholder="Ex: /content/capa.jpg (Dê preferência a JPG/PNG)", description="🖼️ Arquivo Imagem:", layout=widgets.Layout(width='95%'))

btn_executar_acao = widgets.Button(description="⚡ CONFIGURAR OPERAÇÃO NA MATRIX", layout=widgets.Layout(width='95%', height='45px'))
btn_executar_acao.add_class('matrix-btn-gui')

output_console_gui = widgets.Output(layout=widgets.Layout(width='95%', max_height='400px', border='1px solid #00FF00', overflow_y='auto'))

# --- FUNÇÃO AUXILIAR PARA GERAR O NOVO CAMINHO ---
def obter_caminho_saida(target):
    p = pathlib.Path(target)
    # Exemplo: /content/musica.mp3 -> /content/musica_MODIFIED.mp3
    return str(p.parent / f"{p.stem}_MODIFIED{p.suffix}")

# --- FUNÇÕES CORE ---
def run_visualizar(target):
    print("🟩 Coletando logs estruturais da mídia...\n")
    if not os.path.exists("/usr/bin/mediainfo"):
        subprocess.run("apt-get install mediainfo -y >/dev/null 2>&1", shell=True)
    res = subprocess.run(f'mediainfo "{target}"', shell=True, stdout=subprocess.PIPE, text=True)
    print(res.stdout if res.stdout.strip() else "[*] Sem metadados estruturados de leitura direta.")

def run_limpar(target):
    print("🧹 Iniciando expurgo completo e reconstrução em nova cópia binária...")
    ext = pathlib.Path(target).suffix.lower()
    saida_final = obter_caminho_saida(target)

    if ext in ['.jpg', '.jpeg', '.png']:
        img = Image.open(target)
        data = list(img.getdata())
        clean_img = Image.new(img.mode, img.size)
        clean_img.putdata(data)
        clean_img.save(saida_final)
    else:
        cmd = f'ffmpeg -y -i "{target}" -map_metadata -1 -c copy "{saida_final}" >/dev/null 2>&1'
        subprocess.run(cmd, shell=True)

    if os.path.exists(saida_final):
        print(f"[+] Sucesso! Nova cópia purificada gerada em:\n👉 {saida_final}")
    else:
        print("❌ Erro ao purificar metadados deste container.")

def run_editar(target, title, artist, year, genre, album):
    print("✏️ Reinjetando blocos de identificação e gerando nova cópia...")
    saida_final = obter_caminho_saida(target)

    meta_args = ""
    if title: meta_args += f' -metadata title="{title}"'
    if artist: meta_args += f' -metadata artist="{artist}"'
    if year: meta_args += f' -metadata date="{year}"'
    if genre: meta_args += f' -metadata genre="{genre}"'
    if album: meta_args += f' -metadata album="{album}"'

    if not meta_args:
        print("⚠️ Cancelado: Nenhum campo de edição foi preenchido.")
        return

    cmd = f'ffmpeg -y -i "{target}"{meta_args} -c copy "{saida_final}" >/dev/null 2>&1'
    subprocess.run(cmd, shell=True)

    if os.path.exists(saida_final):
        print(f"[+] Código injetado com sucesso! Arquivo modificado criado em:\n👉 {saida_final}")
    else:
        print("❌ Falha na reinjeção dos metadados.")

def run_capa(target, cover):
    if not os.path.exists(cover):
        print(f"❌ [ERRO] Imagem da capa não localizada: {cover}")
        return

    print("🖼️ Vinculando arte visual ao container de mídia em nova cópia...")
    ext = pathlib.Path(target).suffix.lower()
    saida_final = obter_caminho_saida(target)

    if ext in ['.mp3', '.m4a', '.flac']:
        cmd = f'ffmpeg -y -i "{target}" -i "{cover}" -map 0:0 -map 1:0 -c copy -id3v2_version 3 -metadata:s:v title="Album cover" -metadata:s:v comment="Cover (front)" "{saida_final}" >/dev/null 2>&1'
    elif ext in ['.mp4', '.mkv']:
        cmd = f'ffmpeg -y -i "{target}" -i "{cover}" -map 0 -map 1 -c copy -disposition:v:1 attached_pic "{saida_final}" >/dev/null 2>&1'
    else:
        print(f"❌ O formato {ext} não aceita acoplamento de capa por este painel.")
        return

    subprocess.run(cmd, shell=True)
    if os.path.exists(saida_final):
        print(f"[+] Sucesso! Arte visual injetada. Arquivo com Thumbnail criado em:\n👉 {saida_final}")
    else:
        print("❌ Ocorreu um erro no processo de acoplamento do FFmpeg.")

# --- CONTROLADOR CENTRAL ---
def orquestrador_matrix(b):
    with output_console_gui:
        clear_output()

        operacao = drp_operacao.value
        target = txt_target_file.value.strip()
        title = txt_edit_title.value.strip()
        artist = txt_edit_artist.value.strip()
        year = txt_edit_year.value.strip()
        genre = txt_edit_genre.value.strip()
        album = txt_edit_album.value.strip()
        cover = txt_cover_file.value.strip()

        if not target or not os.path.exists(target):
            print(f"❌ [ERRO] Arquivo de mídia não especificado ou inválido na Matrix: '{target}'")
            return

        print(f"🟩 OPERAÇÃO ATIVADA: {operacao.upper()}")
        print("==========================================================\n")

        if operacao == "visualizar":
            run_visualizar(target)
        elif operacao == "limpar":
            run_limpar(target)
        elif operacao == "editar":
            run_editar(target, title, artist, year, genre, album)
        elif operacao == "capa":
            run_capa(target, cover)

        print("\n==========================================================")

btn_executar_acao.on_click(orquestrador_matrix)

# --- RENDERIZAÇÃO DO LAYOUT ---
display(widgets.VBox([
    drp_operacao,
    txt_target_file,
    lbl_edicao,
    txt_edit_title,
    txt_edit_artist,
    txt_edit_year,
    txt_edit_genre,
    txt_edit_album,
    lbl_capa,
    txt_cover_file,
    widgets.HTML("<br>"),
    btn_executar_acao,
    output_console_gui
]))
```

```python
#@title <b><img src="https://upload.wikimedia.org/wikipedia/commons/thumb/1/19/MediaInfo_Logo.svg/960px-MediaInfo_Logo.svg.png?utm_source=es.wikipedia.org&utm_campaign=index&utm_content=thumbnail" width="22" height="22" style="vertical-align: middle; margin-right: 8px;"> MEDIAINFO</b> { display-mode: "form" }
import os, subprocess
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

# Injeção estática Matrix Premium calibrada no Verde Padrão
estilo_matrix_gui = widgets.HTML("""
<style>
    .output_subarea, .output_wrapper, .jupyter-widgets { background-color: #000000 !important; font-family: 'Courier New', monospace !important; }
    .matrix-title-gui { color: #FFFFFF !important; font-family: 'Courier New', monospace; font-weight: bold; font-size: 14px; text-shadow: 0 0 5px #FFFFFF; }
    .widget-label { color: #00FF00 !important; font-family: 'Courier New', monospace; font-weight: bold; min-width: 150px !important; display: inline-block; }
    .widget-text input { background-color: #000000 !important; color: #00FF00 !important; border: 2px solid #00FF00 !important; font-family: 'Courier New', monospace; font-weight: bold; }
    .widget-checkbox { color: #00FF00 !important; font-family: 'Courier New', monospace; font-weight: bold; }
    .matrix-btn-gui { background-color: #005f00 !important; color: #00FF00 !important; border: 2px solid #00FF00 !important; font-family: 'Courier New', monospace !important; font-weight: bold !important; text-shadow: 0 0 5px #00FF00; }
</style>
<div class="matrix-title-gui">🎛️ CENTRAL GRAPHICAL MATRIX — MEDIAINFO EXTRACTION INTERFACE</div>
<hr style="border: 1px dashed #00FF00;">
""")
display(estilo_matrix_gui)

# --- ELEMENTOS INTERATIVOS DA UI (IPYWIDGETS) ---
txt_path_to_file = widgets.Text(
    placeholder="Caminho do arquivo (Ex: /content/video.mp4)",
    description="📂 Destino Arquivo:",
    layout=widgets.Layout(width='95%')
)

chk_save_output = widgets.Checkbox(
    value=False,
    description="💾 Despejar relatório em arquivo de texto (.txt)",
    layout=widgets.Layout(width='95%')
)

btn_analisar_media = widgets.Button(
    description='⚡ FAZER LEITURA E EXTRAÇÃO DE METADADOS (MEDIAINFO)',
    layout=widgets.Layout(width='95%', height='45px')
)
btn_analisar_media.add_class('matrix-btn-gui')

output_console_gui = widgets.Output(layout=widgets.Layout(width='95%', max_height='400px', border='1px solid #00FF00', overflow_y='auto'))

def motor_mediainfo(b):
    with output_console_gui:
        clear_output()

        path_to_file = txt_path_to_file.value.strip()
        save_output_to_file = chk_save_output.value

        # Verificação e instalação do pacote binário em background de forma silenciosa
        if not os.path.exists("/usr/bin/mediainfo"):
            print("🛰️ Injetando dependências binárias do MediaInfo no núcleo Linux...")
            subprocess.run("apt-get update -y >/dev/null 2>&1 && apt-get install mediainfo -y >/dev/null 2>&1", shell=True)

        if not os.path.exists(path_to_file):
            print(f"❌ [ERRO] O arquivo especificado não foi detectado na malha: {path_to_file}")
            return

        print("🟩 Extraindo cabeçalhos e metadados estruturais da mídia...\n")
        print("==========================================================")

        try:
            # Captura os dados do MediaInfo diretamente via pipeline estável
            resultado = subprocess.run(f'mediainfo "{path_to_file}"', shell=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
            media_data = resultado.stdout

            # Limpa o caminho absoluto do diretório para simplificar a visualização do nome do arquivo
            diretorio_base = os.path.dirname(path_to_file) + "/"
            if diretorio_base != "/":
                media_data = media_data.replace(diretorio_base, "")

            print(media_data)

            # Lógica corrigida e limpa para exportação do arquivo de texto
            if save_output_to_file:
                caminho_txt = os.path.splitext(path_to_file)[0] + ".txt"
                if os.path.exists(caminho_txt):
                    os.remove(caminho_txt)
                with open(caminho_txt, 'w', encoding='utf-8') as f:
                    f.write(media_data)
                print(f"🟩 Registro salvo com sucesso em: {caminho_txt}")

        except Exception as e:
            print(f"❌ Erro na varredura do MediaInfo: {e}")

        print("==========================================================")

btn_analisar_media.on_click(motor_mediainfo)

# Renderização do painel operacional
display(widgets.VBox([
    txt_path_to_file,
    chk_save_output,
    widgets.HTML("<br>"),
    btn_analisar_media,
    output_console_gui
]))
```
