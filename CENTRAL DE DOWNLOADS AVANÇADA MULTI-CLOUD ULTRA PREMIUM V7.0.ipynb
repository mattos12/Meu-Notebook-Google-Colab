```python
#@title <b><img src="https://avatars.githubusercontent.com/u/24937341?s=200&v=4" width="22" height="22" style="vertical-align: middle; margin-right: 8px;"> RCLONE WEB GUI</b> { display-mode: "form" }
import os, sys, time, subprocess, ipywidgets as widgets
from IPython.display import display, clear_output

# Injeção estática Matrix Premium calibrada no Verde Padrão
estilo_matrix_gui = widgets.HTML("""
<style>
    .output_subarea, .output_wrapper, .jupyter-widgets { background-color: #000000 !important; font-family: 'Courier New', monospace !important; }
    .matrix-title-gui { color: #FFFFFF !important; font-family: 'Courier New', monospace; font-weight: bold; font-size: 14px; text-shadow: 0 0 5px #FFFFFF; }
    .widget-label { color: #00FF00 !important; font-family: 'Courier New', monospace; font-weight: bold; min-width: 140px !important; display: inline-block; }
    .widget-dropdown select, .widget-text input { background-color: #000000 !important; color: #00FF00 !important; border: 2px solid #00FF00 !important; font-family: 'Courier New', monospace; font-weight: bold; }
    .matrix-btn-gui { background-color: #005f00 !important; color: #00FF00 !important; border: 2px solid #00FF00 !important; font-family: 'Courier New', monospace !important; font-weight: bold !important; text-shadow: 0 0 5px #00FF00; }
</style>
<div class="matrix-title-gui">🎛️ CENTRAL GRAPHICAL MATRIX — RELAY TÚNEL INTERATIVO</div>
<hr style="border: 1px dashed #00FF00;">
""")
display(estilo_matrix_gui)

# Caixas gráficas de seleção e parametrização
drp_tipo_tunel = widgets.Dropdown(
    options=[
        ('1 - LocalTunnel (Padrão)', 'localtunnel'),
        ('2 - Cloudflare Tunnel (Estabilizado)', 'cloudflare'),
        ('3 - Ngrok Tunnel (Requer Token)', 'ngrok')
    ],
    value='localtunnel',
    description='📡 Tipo de Túnel:',
    layout=widgets.Layout(width='95%')
)

txt_ngrok_token = widgets.Text(
    placeholder='🔑 Opcional: Cole seu Authtoken do Ngrok aqui apenas se escolher a Opção 3...',
    description='🔑 Token Ngrok:',
    layout=widgets.Layout(width='95%')
)

btn_ativar_gui_cell = widgets.Button(description='🌐 LIGAR RELAY INTERFACE GRÁFICA (RCLONE WEB GUI)', layout=widgets.Layout(width='95%', height='45px'))
btn_ativar_gui_cell.add_class('matrix-btn-gui')

output_console_gui = widgets.Output(layout=widgets.Layout(width='95%', max_height='250px', border='1px solid #333', overflow_y='auto'))

def disparar_rclone_web_gui_isolado(b):
    with output_console_gui:
        clear_output()
        tipo_conexao = drp_tipo_tunel.value
        token_inserido = txt_ngrok_token.value.strip()

        print("💻 Inicializando Servidor Interno Rclone Web GUI...")
        os.system("pkill -f 'rclone rcd'")
        os.system("pkill -f 'lt --port'")
        os.system("pkill -f 'cloudflared'")
        os.system("pkill -f 'ngrok'")
        time.sleep(1)

        # Ativa o motor da GUI com autenticação firme original
        cmd_gui = "rclone rcd --rc-web-gui --rc-user=matrix --rc-pass=matrix --rc-addr=127.0.0.1:5572 --rc-no-auth=false --rc-web-gui-no-open-browser &"
        subprocess.Popen(cmd_gui, shell=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
        time.sleep(2)

        print("\n==========================================================")
        print("🟩 MATRIX GUI ACTIVATED! INICIALIZANDO ROTEAMENTO:")
        print("==========================================================\n")
        print("👉 Usuário: matrix")
        print("👉 Senha:   matrix\n")

        # -----------------------------------------------------------------
        # OPÇÃO 1: LOCALTUNNEL
        # -----------------------------------------------------------------
        if tipo_conexao == 'localtunnel':
            print("📡 Sincronizando canais de rede via LocalTunnel...")
            os.system("npm install -g localtunnel >/dev/null 2>&1")
            processo_tunnel = subprocess.Popen("lt --port 5572", shell=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
            time.sleep(3)
            for linha_log in iter(processo_tunnel.stdout.readline, ''):
                if "url is:" in linha_log:
                    print(f"🔗 LINK GERADO COM SUCESSO -> {linha_log.split('url is:')[-1].strip()}")
                    break

        # -----------------------------------------------------------------
        # OPÇÃO 2: CLOUDFLARE TUNNEL (INSTALAÇÃO CORRIGIDA VIA REPOSITÓRIO)
        # -----------------------------------------------------------------
        elif tipo_conexao == 'cloudflare':
            print("☁️ Vinculando barramento oficial e estável da Cloudflare...")
            if not os.path.exists("/usr/local/bin/cloudflared"):
                os.system("curl -L --output /usr/local/bin/cloudflared https://github.com >/dev/null 2>&1")
                os.system("chmod +x /usr/local/bin/cloudflared")

            log_cf = "/content/cf_gui.log"
            os.system(f"rm -f {log_cf}")
            subprocess.Popen(f"/usr/local/bin/cloudflared tunnel --url http://127.0.0.1:5572 > {log_cf} 2>&1 &", shell=True)

            print("⏳ Capturando link na malha de servidores (Aguarde 6s)...")
            time.sleep(6)

            url_cf_ok = False
            if os.path.exists(log_cf):
                with open(log_cf, "r") as f:
                    for l in f:
                        if "trycloudflare.com" in l:
                            for parte in l.split("https://"):
                                if "trycloudflare.com" in parte:
                                    link_limpo = "https://" + parte.split()[0].strip()
                                    print(f"🔗 LINK GERADO COM SUCESSO -> {link_limpo}")
                                    url_cf_ok = True; break
            if not url_cf_ok:
                print("⚠️ O link demorando a responder. Aguarde 3 segundos e clique no botão verde novamente.")

        # -----------------------------------------------------------------
        # OPÇÃO 3: NGROK TUNNEL (COM INJEÇÃO DE AUTHTOKEN GRÁFICO)
        # -----------------------------------------------------------------
        elif tipo_conexao == 'ngrok':
            if not token_inserido:
                print("❌ [ERRO] Para rodar a rota do Ngrok, você precisa colar o seu Authtoken no campo 'Token Ngrok' acima!")
                print("👉 Pegue seu token grátis criando uma conta rápida em: https://ngrok.com")
                return

            print("🚀 Autenticando credenciais no gateway Ngrok...")
            if subprocess.run("command -v ngrok", shell=True, stdout=subprocess.DEVNULL).returncode != 0:
                os.system("curl -s https://amazonaws.com | tee /etc/apt/trusted.gpg.key >/dev/null")
                os.system('echo "deb https://amazonaws.com buster main" | tee /etc/apt/sources.list.d/ngrok.list >/dev/null')
                os.system("apt-get update -q && apt-get install ngrok -q -y >/dev/null 2>&1")

            os.system(f"ngrok config add-authtoken {token_inserido} >/dev/null 2>&1")
            subprocess.Popen("ngrok http 5572 > /dev/null 2>&1 &", shell=True)
            time.sleep(4)

            try:
                import urllib.request, json
                res = urllib.request.urlopen("http://127.0.0")
                dados = json.loads(res.read().decode())
                url_ngrok = dados['tunnels'][0]['public_url']
                print(f"🔗 LINK GERADO COM SUCESSO -> {url_ngrok}")
            except Exception as e:
                print("⚠️ Falha ao ler api do Ngrok. Verifique se o token colado está correto.")

        print("\n==========================================================")

btn_ativar_gui_cell.on_click(disparar_rclone_web_gui_isolado)

display(widgets.VBox([
    drp_tipo_tunel,
    txt_ngrok_token,
    widgets.HTML("<br>"),
    btn_ativar_gui_cell,
    output_console_gui
]))
```
